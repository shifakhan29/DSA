# Climbing Stairs - LeetCode 70

Given `n` stairs, you can climb either 1 or 2 steps at a time. Return the number of distinct ways to reach the top.

- A way to climb `n` stairs can finish with:
  - A 1-step climb from `n - 1`
  - A 2-step climb from `n - 2`
- Recurrence: `ways(n) = ways(n - 1) + ways(n - 2)`
- Base cases: `ways(0) = 1`, `ways(1) = 1`

This recurrence is the Fibonacci pattern shifted by one position.

## Approaches covered

1. Plain recursion
2. Top-down recursion with memoization
3. Bottom-up dynamic programming / tabulation
4. Bottom-up DP with constant space

## Approach 1: Plain Recursion

The function directly follows the climbing-stairs recurrence.

- To reach `n`, first reach `n - 1` or `n - 2`
- Base cases: `solve(0) = 1` and `solve(1) = 1`
- Transition: `solve(n) = solve(n - 1) + solve(n - 2)`

The same smaller subproblems are recalculated repeatedly through overlapping recursive branches.

- Time: `O(2^n)` in the worst case
- Auxiliary space: `O(n)` due to recursion-stack depth

````python
//using recursion
class Solution {
public:
    int solve(int n){
        if(n == 0 || n == 1)
            return 1;

        int one_step = solve(n-1);
        int two_step = solve(n-2);

        return one_step + two_step;
    }

    int climbStairs(int n) {
        return solve(n);
        
    }
};
````

## Approach 2: Recursion + Memoization (Top-Down DP)

We calculate the same recurrence as plain recursion, but store each result when it is first calculated.

**State definition:**
- `memo[n]` stores the number of ways to climb `n` stairs
- `memo[n] == -1` means the state has not been calculated yet

**Initialization and transition:**
- Base cases: `memo[0] = 1` and `memo[1] = 1`
- Transition: `memo[n] = solve(n - 1, memo) + solve(n - 2, memo)`
- Direction: **Top-down** - larger states are broken into smaller states

- Time: `O(n)`
- Auxiliary space: `O(n)` for the memo table and recursion stack

````python
//using memoization
class Solution {
public:
    int solve(int n,vector<int>&memo){

        if(n==0 || n == 1)
           return 1;
        if(memo[n]!=-1)
           return memo[n];

        int one_step = solve(n-1, memo);
        int two_step = solve(n-2,memo);

        return memo[n] = one_step + two_step;
    }

    int climbStairs(int n) {
        vector<int>memo(n+1 ,-1);
        return solve(n,memo);
        
    }
};
````

## Approach 3: Tabulation (Bottom-Up DP)

Instead of recursion, we build the number of ways for every stair count from the smallest state to `n`.

**State definition:**
- `tab[i]` stores the number of distinct ways to reach stair `i`

**Initialization and transition:**
- `tab[0] = 1`
- `tab[1] = 1`
- `tab[2] = 2`
- For `i = 3` to `n`: `tab[i] = tab[i - 1] + tab[i - 2]`
- Final answer: `tab[n]`
- Direction: **Bottom-up** - smaller states are calculated before larger states
- Time: `O(n)`
- Auxiliary space: `O(n)`

````python
// using tabulation
class Solution {
public:
    int climbStairs(int n) {
        if (n == 0 || n == 1)
            return 1;
        vector<int> tab(n + 1, -1);
        tab[0] = 1;
        tab[1] = 1;
        tab[2] = 2;

        for (int i = 3; i <= n; i++) {
            tab[i] = tab[i - 1] + tab[i - 2];
        }
        return tab[n];
    }
};
````

## Approach 4: Bottom-Up DP with Constant Space

We only need the previous two answers, so we do not need to store the complete DP table.

**Variables used:**
- `first` stores the number of ways for `i - 2`
- `second` stores the number of ways for `i - 1`
- `next` stores `first + second`, which is the answer for `i`

**Update pattern for each iteration:**
```text
next = first + second
first = second
second = next
```

- Base cases: `n <= 1 -> 1`
- Time: `O(n)`
- Auxiliary space: `O(1)`

````python
// using constant space
class Solution {
public:
    int climbStairs(int n) {
        if (n == 0 || n == 1)
            return 1;

        int first = 1;
        int second = 1;
        for (int i = 2; i <= n; i++) {
            int next = first + second;
            first = second;
            second = next;
        }
        return second; // TC:O(n) || SC:O(1)
    }
};
````
