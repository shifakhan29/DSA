# Fibonacci Number — LeetCode 509

Given `n`, return the `n`th Fibonacci number.

- Fibonacci definition: `fib(0) = 0`, `fib(1) = 1`
- Recurrence: `fib(n) = fib(n - 1) + fib(n - 2)` for `n >= 2`


## Approaches covered

1. Top-down recursion with memoization
2. Bottom-up dynamic programming / tabulation
3. Bottom-up DP with constant space
4. Plain recursion

## Approach 1: Recursion + Memoization (Top-Down DP)

We try to calculate `fib(n)` recursively, but each subproblem is stored in a `dp` array so that it is not calculated repeatedly.

**State definition:**
- `dp[n]` stores the Fibonacci value for `n`
- `dp[n] == -1` means the value has not been calculated yet

**Why memoization works:**
Every call to `solve(n, dp)` follows the recurrence `solve(n) = solve(n - 1) + solve(n - 2)`. Once `solve(n)` is evaluated, its result is reused whenever `n` appears again.

- Base cases: `n <= 1 -> n`
- Transition: `dp[n] = solve(n - 1) + solve(n - 2)`
- Direction: **Top-down** — solve larger states by breaking them into smaller states
- Time: `O(n)`
- Auxiliary space: `O(n)` for the DP array plus recursion stack
- Overall space: `O(n)`

````python
//using memoization
class Solution{
    public:
    
    int solve(int n, vector<int>&dp){
        if(n<=1)
            return n;
        if(dp[n] != -1)
return dp[n];
        
        return dp[n] = solve(n-1 , dp) + solve(n-2, dp);
        
    }
    int fib(int n){
        vector<int>dp(n+1,-1);
        return solve(n,dp);//TC:O(n) || SC:O(n) -->top down approach
    }
};
````

## Approach 2: Tabulation (Bottom-Up DP)

Instead of using recursion, we first calculate every Fibonacci value from the smallest subproblem to the required answer.

**State definition:**
- `tab[i]` stores the `i`th Fibonacci number

**Initialization and transition:**
- `tab[0] = 0`
- `tab[1] = 1`
- For `i = 2` to `n`: `tab[i] = tab[i - 1] + tab[i - 2]`
- Final answer: `tab[n]`

- Base cases are handled directly before filling the table
- Direction: **Bottom-up** — smaller states are calculated before larger states
- Time: `O(n)`
- Auxiliary space: `O(n)` because the complete Fibonacci table is stored

````python
//using tabulation
class Solution{
    public:
    int fib(int n){
        if(n<=1)
        return n;
        vector<int>tab(n+1,-1);
        tab[0] = 0;
        tab[1] = 1;

        for(int i = 2 ; i<=n;i++){
        
            tab[i] = tab[i-1] + tab[i-2];
        }
    return tab[n];//TC:O(n)  || SC: O(n)-->bottom up approach
    }
};
````

## Approach 3: Bottom-Up DP with Constant Space

Each Fibonacci number depends only on the previous two values. Therefore, we do not need to store the complete DP table.

**Variables used:**
- `a` stores `fib(i - 2)`
- `b` stores `fib(i - 1)`
- `c` stores `a + b`, which is `fib(i)`

**Update pattern for each iteration:**
```text
c = a + b
a = b
b = c
```

- Base cases: `n <= 1 -> n`
- Time: `O(n)`
- Auxiliary space: `O(1)`
- This is the most space-efficient approach for the Fibonacci recurrence

````python
//using consatnt space
class Solution {
public:
    int fib(int n) {
        if (n <= 1)
            return n;
        int a = 0;
        int b = 1;

        for (int i = 2; i <= n; i++) {
            int c = a + b;
            a = b;
            b = c;
        }
        return b;
    }
};
````

## Approach 4: Plain Recursion (Brute Force)

This implementation directly follows the mathematical Fibonacci recurrence without memoization or a DP table.

- Base cases: `n <= 1 -> n`
- Recursive transition: `fib(n) = fib(n - 1) + fib(n - 2)`

**Why it is slower:**
The recursive calls recompute the same subproblems many times. For example, `fib(n - 1)` and `fib(n - 2)` may later repeat overlapping branches of the recursion tree.

- Time: `O(2^n)` in the worst case
- Auxiliary space: `O(n)` due to the maximum recursion-stack depth
- This approach is useful for understanding the recurrence, but it should not be used for large values of `n`

````python
//using recursion
class Solution {
public:
    int fib(int n) {
        if(n<=1)
        return n;
        return fib(n-1) + fib(n-2);
        
    }
};
````
