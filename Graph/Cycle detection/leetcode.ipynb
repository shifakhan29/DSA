# 1. Number of Provinces (LeetCode - 547)

**Input:**
```cpp
isConnected = [[1,1,0],
               [1,1,0],
               [0,0,1]]
```

**Output:** `2`

## Approach 1: BFS

### Algorithm
1. Initialize a `visited` array of size `n` with `false`.
2. Traverse all vertices from `0` to `n-1`.
3. If a vertex is unvisited:
   - Increment `count` (new province found).
   - Call BFS from that vertex.
4. In BFS:
   - Mark the starting vertex as visited and push it into the queue.
   - While the queue is not empty:
     - Remove the front vertex `u`.
     - Traverse all vertices `v`.
     - If `v` is unvisited and `isConnected[u][v] == 1`:
       - Mark `v` visited.
       - Push `v` into the queue.
5. Return `count`.

### Code
```cpp
// Using BFS

class Solution {
public:
    void BFS(vector<vector<int>>& isConnected, vector<bool>& visited, int u) {
        queue<int> q;
        q.push(u);
        visited[u] = true;

        while (!q.empty()) {
            int curr = q.front();
            q.pop();

            for (int v = 0; v < isConnected.size(); v++) {
                if (!visited[v] && isConnected[curr][v] == 1) {
                    visited[v] = true;
                    q.push(v);
                }
            }
        }
    }

    int findCircleNum(vector<vector<int>>& isConnected) {
        int n = isConnected.size();
        vector<bool> visited(n, false);
        int count = 0;

        for (int i = 0; i < n; i++) {
            if (!visited[i]) {
                BFS(isConnected, visited, i);
                count++;
            }
        }

        return count;
    }
};
```

**Complexity Analysis:**
- **TC:** `O(n²)`
- **SC:** `O(n)` — visited array + queue.

## Approach 2: DFS

### Algorithm
1. Initialize a `visited` array of size `n` with `false`.
2. Traverse all vertices from `0` to `n-1`.
3. If a vertex is unvisited:
   - Increment `count` (new province found).
   - Call DFS from that vertex.
4. In DFS:
   - Mark the current vertex `u` as visited.
   - Traverse all vertices `v`.
   - If `v` is unvisited and `isConnected[u][v] == 1`, recursively call DFS on `v`.
5. Return `count`.

### Code
```cpp
// Using DFS

class Solution {
public:
    void DFS(vector<vector<int>>& isConnected, vector<bool>& visited, int u) {
        visited[u] = true;

        for (int v = 0; v < isConnected.size(); v++) {
            if (!visited[v] && isConnected[u][v] == 1) {
                DFS(isConnected, visited, v);
            }
        }
    }

    int findCircleNum(vector<vector<int>>& isConnected) {
        int n = isConnected.size();
        int count = 0;
        vector<bool> visited(n, false);

        for (int i = 0; i < n; i++) {
            if (!visited[i]) {
                DFS(isConnected, visited, i);
                count++;
            }
        }

        return count;
    }
};
```

**Complexity Analysis:**
- **TC:** `O(n²)`
- **SC:** `O(n)` — visited array + recursion stack.

---

### Key Points
- A province represents a **connected component**.
- Every new unvisited vertex starts a new province.
- BFS uses a **Queue**.
- DFS uses **Recursion / Stack**.
- Both approaches give the same result.

# 2. Course Schedule I (Leetcode - 207)
## Using BFS : 
* ### Algorithm
1. Build Adjacency List
   - For every prerequisites [a,b], create a directed edge b->a.
   - Course b must be completed course a.
2. Calculate Indegree
   - Create an indegree array of size n, initialized with 0.
   - Increment indegree[0] for every incoming edge. 
3. Initialise Queue
   - Push all vertices having indegree = 0 into the queue.
   - These courses have no prerequisites.
4. Process BFS
   - While the queue is not empty:
   1. Extract the front vertex u ans pop it.
   2. Increment count (one course processed).
   3. Decrease indegree[v] by 1.
   4. If indegree[v] == 0, push v into the queue..
5. Cycle Detection
  - If count == numCourses, return true(all courses can be completed).
  - Otherwise, return false (a directed cycle exists).
 ### Complexity Analysis
 * Time Complexity : O(V+E)
 * Space Complexity : O(V+E)

### Code

```cpp
//using topological sort

class Solution{
    public:
        bool canFinish(int numCourses, vector<vector<int>>& prerequisites){
            int n = numCouses.size();
            int p = prerequisites.size();
            unordered_map<int,vector<int>>adj;
            vector<int>indegree(n,0);
            //fill the in degree
            for(int i=0; i<p ; i++){
                int a = prerequisites[i][0];
                int b = prerequisites[i][1];// arrow  b-->a
                adj[b].push_back(a);
                //arrow is going from b -->a
                indegree[a]++;
            }
            return topoSortCheck(adj, indegree, numCourses);
        }

        bool topoSortCheck(unordered_map<int,vector<int>>&adj, vector<int>&indegree, int numCourses){
            queue<int>q;
            int count = 0 ; 
            for(int i =0; i < numCourses.size() ; i++){
                if(indegree[i]==0){
                    q.push(i);
                }
            }
            while(!q.empty()){
                int u =q.front();
                q.pop();

                count++; // processed one course
                for(int &v : adj[u]){
                    indegree[v]--;
                    if(indegree[v] == 0)
                    {
                        q.push(v);
                    }
                }
            }
            return (count == numCourses);
```
## DFS: Cycle Detection (Course Schedule)

### Algorithm
1. Create a directed adjacency list (`b → a`).
2. Initialize `visited[]` and `inRecursion[]` with `false`.
3. Traverse all vertices:
   - If a vertex is unvisited, call DFS.
4. In DFS:
   - Mark the current node as `visited` and `inRecursion`.
   - Traverse all adjacent vertices:
     - If unvisited, recursively call DFS.
     - If already present in `inRecursion`, a cycle is detected; return `true`.
   - Backtrack by setting `inRecursion[u] = false`.
   - Return `false` if no cycle is found.
5. If any DFS call detects a cycle, return `false` (courses cannot be completed).
6. If all vertices are processed without detecting a cycle, return `true`.

### Complexity Analysis
- **TC:** O(V + E)
- **SC:** O(V + E) 
```cpp
class Solution {
public:
   bool hasCycleDfs(unordered_map<int,vector<int>>&adj, 
                    vector<bool>&visited, 
                    vector<bool>&inRecursion, 
                    int u){
    visited[u] = true;
    inRecursion[u] = true;
    for( int &v : adj[u]){
        if(!visited[v] && hasCycleDfs(adj, visited, inRecursion, v)){
            return true;
        }
        else if(inRecursion[v] == true)
        {
            return true;
        }
    }
    //backtrack
    inRecursion[u] = false;
    
    return false;
   }

    bool canFinish(int numCourses, vector<vector<int>>& prerequisites) {
        int n = numCourses;
        int p = prerequisites.size();

        unordered_map<int,vector<int>>adj;
        vector<bool>visited(n,false);
        vector<bool>inRecursion(n,false);

        for(int i =0 ; i< p; i++){
            int a = prerequisites[i][0];
            int b = prerequisites[i][1];
            adj[b].push_back(a);
        }


        for(int i = 0; i<n ;i++){
            if(!visited[i] && hasCycleDfs(adj, visited, inRecursion,i))
            {
                return false;//if cycle exists return false
            }
        }
        return true;//cycle does not exists 
    }
};//TC: O(V+E) || SC: O(V+E)
```

## BFS - Topological sort + cycle detection
* ### Algorithm

1. Build Adjacency List.(b-->a)
   - For each prerequisite [a,b], createa directed edge b -> a.
   - Course b must be completed before course a. 
2. Calculate indegree.
   - maintain an indegree array of size n, initialized to 0.
   - Increment indegree[a] for every incoming edge.
3. Initialize Queue.
   - Push all courses having indegree = 0 into the queue.
   - These courses have no prerequisites.
4. Process BFS.
   - While the queue is not empty:
    1. Extract the front course.
    2. Add it to the result array.
    3. Traverse its adjacent courses.
    4. Decrease their indegree by 1.
    5. If any course's indegree becomes  0, push it into the queue.
5. Check Cycle. 
   - If ans.size() == numCourses, return ans.
   - Otherwise, a cycle exists, so return {}

### Complexity Analysis

- Time Complexity: O(V + E)
- Space Complexity: O(V + E)
  1. Adjacency list: O(V + E) 
  2. Indegree array: O(V)
  3. Queue: O(V)
  4. Result array: O(V)
 
```cpp
//using topological sort
class Solution{
public:
    vector<int> findOrder(int numCourses, vector<vector<int>>& prerequisites) {
        int n = numCourses;
        int p = prerequisites.size();
        vector<int>ans;
        vector<int>indegree(n,0);
        unordered_map<int, vector<int>>adj;
        for(int i=0 ; i<p; i++){
            int a = prerequisites[i][0];
            int b = prerequisites[i][1];
            adj[b].push_back(a);
            indegree[a]++;

        }
        queue<int>q;
        for(int i =0; i<n; i++){
            if(indegree[i]==0)
            q.push(i);
        }
        while(!q.empty()){
            int u = q.front();
            q.pop();

            ans.push_back(u);

            for(int &v : adj[u]){
                indegree[v]--;
                if(indegree[v]==0)
                {
                    q.push(v);
                }
            }
        }
        if (ans.size()!=n)
        return {};
        return ans;
    }
};
```
## DFS: Topological Sort + Cycle Detection

### Algorithm
1. Create an adjacency list (`b → a`).
2. Initialize `visited[]`, `inRecursion[]` and a stack.
3. Traverse all vertices and call DFS for every unvisited vertex.
4. In DFS:
   - Mark `u` as visited and add it to `inRecursion`.
   - Traverse all adjacent vertices `v`:
     - If `v` is unvisited, recursively call DFS.
     - If `inRecursion[v] == true`, a cycle exists; return `true`.
   - After processing all neighbours:
     - Set `inRecursion[u] = false`.
     - Push `u` into the stack.
   - Return `false` if no cycle is detected.
5. If DFS detects a cycle, return an empty array `{}`.
6. Otherwise, pop all elements from the stack and store them in the answer.
7. Return the topological ordering.

### Complexity Analysis
- **TC:** O(V + E)
- **SC:** O(V + E)
  
### Code(DFS)

```cpp
class Solution {
public:
    bool hasCycleDfs(unordered_map<int, vector<int>>& adj,
                     vector<bool>& inRecursion, vector<bool>& visited,
                     stack<int>& st, int u) {
        visited[u] = true;
        inRecursion[u] = true;

        for (int& v : adj[u]) {
            if (visited[v] == false &&
                hasCycleDfs(adj, inRecursion, visited, st, v)) {

                return true;
            } else if (inRecursion[v] == true) {
                return true;
            }
        }
        // Backtracking
        inRecursion[u] = false;
        st.push(u); // topological ordering
        return false;
    }

    vector<int> findOrder(int numCourses, vector<vector<int>>& prerequisites) {
        int n = numCourses;
        int p = prerequisites.size();

        unordered_map<int, vector<int>> adj;
        vector<bool> inRecursion(n, false);
        vector<bool> visited(n, false);
        stack<int> st;

        //1. make graph
        for (int i = 0; i < p; i++) {
            int a = prerequisites[i][0];
            int b = prerequisites[i][1]; // direc of arrow == b --> a
            adj[b].push_back(a);
        }
        vector<int> ans;
        //2. func call for dfs
        for (int i = 0; i < n; i++) {
            if (!visited[i] && hasCycleDfs(adj, inRecursion, visited, st, i)) {
                return {};
            }
        }
        //3. Store the ans 
        while (!st.empty()) {
            ans.push_back(st.top());
            st.pop();
        }
        return ans;//TC : O(V + E) || SC : O(V + E)
    }
};
```

# 4. Cycle Detection in Undirected Graph using DSU(GFG)

## Approach

- Initially, every vertex belongs to a separate set.
- Use `find()` to get the ultimate parent (leader).
- Use `Union()` to combine two different sets.

### Algorithm

1. Initialize `parent[i] = i` and `rank[i] = 0`.
2. Traverse every edge `(u, v)`.
3. Process each undirected edge only once using `if(u < v)`.
4. Find both ultimate parents:
   ```cpp
   parent_u = find(u, parent);
   parent_v = find(v, parent);
   ```
5. If both parents are the same:
   - Already belong to the same set.
   - **Cycle detected → return true.**
6. Otherwise:
   - Belong to different sets.
   - Perform `Union(u, v)`.
7. If traversal finishes, return `false`.

## Core Logic

```cpp
if(parent_u == parent_v)
    return true;  // Cycle

Union(u, v, parent, rank); // Combine sets
```

## Complexity Analysis

Let V = vertices and E = edges.

| Operation | Complexity |
|---|---|
| Initialization | O(V) |
| Graph traversal | O(V + E) |
| Find + Union | O(α(V)) amortized per operation |
| **Total TC** | **O(V + E α(V))** |
| **SC** | **O(V)** |

Here, α(V) = Inverse Ackermann function (practically constant).

### Remember

**Same parent → Cycle**

**Different parent → Union**

*Applicable to undirected graphs.*
### Code:
```cpp
class Solution {
public:

    // Find with Path Compression
    int find(int u, vector<int>& parent) {

        if(u == parent[u])
            return u;

        return parent[u] = find(parent[u], parent);
    }

    // Union by Rank
    void Union(int u, int v, vector<int>& parent,
               vector<int>& rank) {

        int u_parent = find(u, parent);
        int v_parent = find(v, parent);

        if(u_parent == v_parent)
            return;

        if(rank[u_parent] > rank[v_parent]) {
            parent[v_parent] = u_parent;
        }

        else if(rank[u_parent] < rank[v_parent]) {
            parent[u_parent] = v_parent;
        }

        else {
            parent[u_parent] = v_parent;
            rank[v_parent]++;
        }
    }

    bool detectCycle(int V, vector<int> adj[]) {

        vector<int> parent(V);
        vector<int> rank(V, 0);

        // Initialization
        for(int i = 0; i < V; i++) {
            parent[i] = i;
        }

        // Traverse graph
        for(int u = 0; u < V; u++) {

            for(int &v : adj[u]) {

                // Process each undirected edge only once
                if(u < v) {

                    int parent_u = find(u, parent);
                    int parent_v = find(v, parent);

                    // Both vertices already belong to same set
                    if(parent_u == parent_v) {
                        return true;
                    }

                    // Combine both sets
                    Union(u, v, parent, rank);
                }
            }
        }

        return false;
    }
};
```

# 4. Leetcode - 990 | Satisfiability of Equality Equations

**## Approach**

- Use **DSU (Union by Rank + Path Compression)** to group equal variables.
- Initially, every character (`a-z`) belongs to a separate set.
- Process all `==` equations first, then check `!=` equations.

**## Algorithm**

1. Initialize `parent[26]` and `rank[26] = 0`.
2. **First Pass – Equality (`==`)**
   - Find the ultimate parents of both variables.
   - If parents are different, perform `Union()`.
   - Both variables now belong to the same set.

3. **Second Pass – Inequality (`!=`)**
   - Find the ultimate parents of both variables.
   - If both parents are the same → **Contradiction → return false.**
   - Otherwise, continue.

4. If no contradiction exists, return `true`.

**### Example**

```text
a == b
e == c
b == c
a != e
```

- `a == b` → Union(a, b).
- `e == c` → Union(e, c).
- `b == c` → Union(a, e).
- Now `a` and `e` belong to the same set.
- `a != e` → Same parent → **return false**.

**## Core Logic**

```cpp
// First pass: Process equality
if(s[1] == '=')
    Union(s[0]-'a', s[3]-'a', parent, rank);

// Second pass: Check inequality
if(s[1] == '!') {
    if(find(s[0]-'a', parent) == find(s[3]-'a', parent))
        return false;
}
```

**## Complexity Analysis**

Let `n` = number of equations.

| Operation | Complexity |
|---|---|
| Initialization | O(26) |
| Union + Find | O(n α(26)) |
| **Total TC** | **O(n)** |
| **SC** | **O(26) = O(1)** |

Here, α = Inverse Ackermann function (practically constant).

**## Remember**

- `a == b` → Union(a, b).
- `a != b` → Check ultimate parents.
- Same parent in `!=` → Contradiction → `false`.
- Different parent → Continue.

**First UNION all equalities → Then CHECK all inequalities.**

**### Code**

```cpp
class Solution {
public:
    int find(int i, vector<int>& parent) {
        if(i == parent[i])
            return i;

        return parent[i] = find(parent[i], parent);
    }

    void Union(int x, int y, vector<int>& parent, vector<int>& rank) {
        int x_parent = find(x, parent);
        int y_parent = find(y, parent);

        if(x_parent == y_parent)
            return;

        if(rank[x_parent] > rank[y_parent])
            parent[y_parent] = x_parent;

        else if(rank[x_parent] < rank[y_parent])
            parent[x_parent] = y_parent;

        else {
            parent[x_parent] = y_parent;
            rank[y_parent]++;
        }
    }

    bool equationsPossible(vector<string>& equations) {
        vector<int> parent(26);
        vector<int> rank(26, 0);

        for(int i = 0; i < 26; i++)
            parent[i] = i;

        // First pass: Process ==
        for(string& s : equations) {
            if(s[1] == '=')
                Union(s[0]-'a', s[3]-'a', parent, rank);
        }

        // Second pass: Check !=
        for(string& s : equations) {
            if(s[1] == '!') {
                int first = find(s[0]-'a', parent);
                int last = find(s[3]-'a', parent);

                if(first == last)
                    return false;
            }
        }

        return true;
    }
};
```
