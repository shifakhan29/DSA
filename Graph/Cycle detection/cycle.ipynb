## 1. Cycle Detection in an undirected graph

### DFS : 

## Algorithm 

1. Initialize a `visited` array of size `n` with `false`.
2. Traverse all vertices from `0` to `n-1`.
3. If a vertex is unvisited, call DFS with `parent = -1`.
4. In DFS:
   - Mark current vertex `u` as visited.
   - Traverse all adjacent vertices `v`.
   - If `v == parent`, skip it.
   - If `v` is already visited, return `true` (cycle detected).
   - Otherwise, recursively call DFS on `v`, passing `u` as its parent.
5. If DFS detects a cycle, return `true`.
6. If all components are traversed without finding a cycle, return `false`.

### Complexity Analysis
- **TC:** `O(V + E)`
- **SC:** `O(V)` — visited array + recursion stack.


```python
//cycle detection in undirected graph using dfs
class Solution {
public:
    bool isCycleDfs(vector<vector<int>> &adj,int u,vector<bool> &visited,int parent) {

        visited[u] = true;

        for (int v : adj[u]) {

            if (v == parent)
                continue;

            if (visited[v])
                return true;

            if (isCycleDfs(adj, v, visited, u))
                return true;
        }

        return false;
    }

    bool isCycle(vector<vector<int>> &adj) {

        int n = adj.size();
        vector<bool> visited(n, false);

        for (int i = 0; i < n; i++) {

            if (!visited[i]) {
                if (isCycleDfs(adj, i, visited, -1))
                    return true;
            }
        }

        return false;
    }
};
```

## 2. Cycle Detection in an undirected graph

### BFS :

## Algorithm — Cycle Detection in Undirected Graph (BFS)

1. Initialize a `visited` array of size `n` with `false`.
2. Traverse all vertices from `0` to `n-1`.
3. If a vertex is unvisited, call BFS from that vertex.
4. In BFS:
   - Initialize a queue storing `{node, parent}`.
   - Push `{u, -1}` into the queue and mark `u` as visited.
5. While the queue is not empty:
   - Extract `{node, parent}` from the queue.
   - Traverse all adjacent vertices `v`.
   - If `v` is unvisited:
     - Mark `v` as visited.
     - Push `{v, node}` into the queue.
   - Else if `v != parent`, return `true` (cycle detected).
6. If BFS detects a cycle, return `true`.
7. If all components are traversed without finding a cycle, return `false`.

### Complexity Analysis
- **TC:** `O(V + E)`
- **SC:** `O(V)` — visited array + queue.
  

```python
//cycle detection in an undirected graph using bfs
class mk Solution {
	public:
	bool isCycleBfs(vector<vector<int>> &adj,int u,vector<bool> &visited) {
		queue<pair<int, int>>q; // {node, parent}
		q.push({u, -1});
		
		visited[u] = true;
		
		while (!q.empty()) {
			auto p = q.front();
			q.pop();
			int node = p.first;
			int parent = p.second;
			
			for (int &v : adj[node]) {
				if (visited[v] != true) {
					visited[v] = true;
					q.push({v, node});
				}
				else if (v != parent) {
					return true;
				}
			}
		}
		return false;
	}
	
	bool isCycle(vector<vector<int>> &adj) {
		
		int n = adj.size();
		vector<bool> visited(n, false);
		
		for (int i = 0; i < n; i++) {
			
			if (!visited[i]) {
				if (isCycleBfs(adj, i, visited))
					return true;
			}
		}
		return false;
	}//TC:O(V + E) || SC:O(V)
};
```

## 3. Cycle Detection in a directed graph

### DFS :

## Algorithm — Cycle Detection in Directed Graph (DFS)

1. Initialize two arrays of size `n`:
   - `visited` → tracks visited vertices.
   - `inRecursion` → tracks vertices currently in the DFS recursion stack.
2. Traverse all vertices from `0` to `n-1`.
3. If a vertex is unvisited, call DFS.
4. In DFS:
   - Mark current vertex `u` as visited.
   - Mark `u` as `true` in `inRecursion`.
   - Traverse all adjacent vertices `v`:
     - If `v` is unvisited, recursively call DFS on `v`.
     - If `inRecursion[v] == true`, a cycle exists → return `true`.
5. After processing all adjacent vertices, mark `inRecursion[u] = false`.
6. If any DFS call detects a cycle, return `true`.
7. If all components are traversed without finding a cycle, return `false`.

### Complexity Analysis
- **TC:** `O(V + E)`
- **SC:** `O(V)` — visited array + recursion stack.


```python
//cycle detection in a directed graph using dfs
class Solution {
	public:
	bool isCyclic(vector<vector<int>> & adj) {
		// code here
		int n = adj.size();
		
		vector<bool> visited(n, false);
		vector<bool> inRecursion(n, false);
		
		for (int i = 0; i <n ; i++) {
			if (!visited[i] && isCycleDFS(adj, i, visited, inRecursion)) {
				return true;
			}
			
		}
		return false;
	}
	bool isCycleDFS(vector<vector<int>> & adj, int u, vector<bool>&visited,
	vector<bool>&inRecursion) {
		visited[u] = true;
		inRecursion[u] = true;
		
		for (int & v : adj[u]) {
			if (visited[v] == false && isCycleDFS(adj, v, visited, inRecursion))
				return true;
			else if (inRecursion[v] == true)
				return true;
		}
		inRecursion[u] = false;
		return false;
	} // TC:O(V + E) || SC:O(V)
};
```

## 4. Cycle Detection in a directed graph

### BFS :

## Algorithm — Cycle Detection in Directed Graph (Kahn's Algorithm / BFS)

1. Initialize an `indegree` array of size `n` with `0`.
2. Calculate indegree:
   - Traverse all vertices and their adjacent vertices.
   - Increment `indegree[v]` for every incoming edge.
3. Initialize Queue:
   - Push all vertices having `indegree = 0` into the queue.
4. Process BFS:
   - Initialize `count = 0`.
   - While the queue is not empty:
     - Remove the front vertex `u`.
     - Increment `count` (processed vertex).
     - Traverse all adjacent vertices `v`.
     - Decrease `indegree[v]` by `1`.
     - If `indegree[v] == 0`, push `v` into the queue.
5. Cycle Detection:
   - If `count != n`, return `true` (cycle exists).
   - Otherwise, return `false` (no cycle).

### Complexity Analysis
- **TC:** `O(V + E)`
- **SC:** `O(V)` — indegree array + queue.

```python
// using Kahn's algorithm for cycle detection in directed graph
//TC: O(V + E) || SC:O(V)

class Solution {
	public:
	bool isCyclic(vector<vector<int>> & adj) {
		int n = adj.size();
		queue<int>q;
		vector<int>indegree(n, 0);
		// 1.store the indegree
		
		for (int i = 0; i<n; i++) {
			for (int & v : adj[i]) {
				indegree[v]++;
			}
		}
		
		// 2.fill the queue
		for (int i = 0; i<n ; i++) {
			if (indegree[i] == 0) {
				q.push(i);
			}
		}
		// 3.normal BFS
		int count = 0; // cycle detect
		while (!q.empty()) {
			int u = q.front();
			q.pop();
			
			count++;//processed vertex
			
			for (int &v : adj[u]) {
				indegree[v]--;
				if (indegree[v] == 0) {
					q.push(v);
					
				}
			}
        }
		return (count != n);
		//TC: O (V+E)  ||  SC:O(V)
	}
};
```
