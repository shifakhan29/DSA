# Euler Path and Euler Circuit

## 1. Euler Path

An Euler Path is a path that uses every edge of the graph exactly once.

- Vertices may be visited multiple times.
- An edge cannot be used twice.
- The starting and ending vertices may be different.

## 2. Euler Circuit

An Euler Circuit is an Euler Path that starts and ends at the same vertex.

- Every Euler Circuit is an Euler Path.
- Every Euler Path is not an Euler Circuit.

## 3. Degree of a Vertex

The degree of a vertex is the number of edges connected to it.

```cpp
degree[i] = adj[i].size();
```

- Even degree: connected to an even number of edges.
- Odd degree: connected to an odd number of edges.

## 4. Undirected Graph Conditions

All vertices having edges must belong to the same connected component.

| Odd-Degree Vertices | Result | Return Value |
|---|---|---|
| 0 | Euler Circuit | 2 |
| 2 | Euler Path, but not Euler Circuit | 1 |
| More than 2 | Neither Euler Path nor Euler Circuit | 0 |

## 5. Approaches

### Step 1: Check Connectivity

1. Find a vertex that has at least one edge.
2. Run DFS from this vertex.
3. Mark every reachable vertex as visited.
4. If another vertex with edges is unvisited, the graph is disconnected.
5. Ignore isolated vertices.

### Step 2: Count Odd-Degree Vertices

```cpp
int oddCount = 0;

for(int i = 0; i < n; i++) {
    if(adj[i].size() % 2 != 0)
        oddCount++;
}
```

### Step 3: Classify the Graph

- `oddCount == 0` → Euler Circuit
- `oddCount == 2` → Euler Path only
- `oddCount > 2` → Neither

## 6. Pseudocode

```text
isEulerCircuit(adj):
    n = number of vertices

    if isConnected(adj) == false:
        return 0

    oddCount = 0

    for i = 0 to n - 1:
        if degree[i] is odd:
            oddCount = oddCount + 1

    if oddCount == 0:
        return 2

    if oddCount == 2:
        return 1

    return 0
```

## 7. Complexity

- DFS connectivity: `O(V + E)`
- Odd-degree count: `O(V)`
- Total time: `O(V + E)`
- Auxiliary space: `O(V)`


````python
class Solution {
public:
    void dfs(vector<vector<int>>& adj, vector<int>& visited, int u) {
        visited[u] = true;

        for(int &v : adj[u]) {
            if(!visited[v])
                dfs(adj, visited, v);
        }
    }

    bool isConnected(vector<vector<int>>& adj) {
        int n = adj.size();
        vector<int> visited(n, false);
        int start = -1;

        // Find a vertex with at least one edge.
        for(int i = 0; i < n; i++) {
            if(adj[i].size() > 0) {
                start = i;
                break;
            }
        }

        // The graph has no edges.
        if(start == -1)
            return true;

        dfs(adj, visited, start);

        // Ignore isolated vertices; check every vertex having edges.
        for(int i = 0; i < n; i++) {
            if(adj[i].size() > 0 && !visited[i])
                return false;
        }

        return true;
    }

    int isEulerCircuit(vector<vector<int>>& adj) {
        int n = adj.size();

        // First condition: graph must be connected.
        if(!isConnected(adj))
            return 0;

        int oddCount = 0;

        // Second condition: count odd-degree vertices.
        for(int i = 0; i < n; i++) {
            if(adj[i].size() % 2 != 0)
                oddCount++;
        }

        if(oddCount == 0)
            return 2; // Euler circuit

        if(oddCount == 2)
            return 1; // Euler path only

        return 0; // Neither
    }
};
````

# Implementation 2: Euler-Circuit-Only Check

This second implementation checks whether every vertex has an even degree.

```text
Euler Circuit exists only when oddCount == 0.
```

This code:

- Does not check connectivity.
- Does not identify Euler paths with exactly two odd vertices.
- Is useful only when the full Euler-circuit code is not required.


````python
class Solution {
public:
    bool isEularCircuitExist(int V, vector<int> adj[]) {
        vector<int> degree(V, 0);

        // Calculate the degree of every vertex.
        for(int i = 0; i < V; i++)
            degree[i] = adj[i].size();

        // An Euler circuit requires every vertex to have even degree.
        for(int i = 0; i < V; i++) {
            if(degree[i] % 2 != 0)
                return false;
        }

        return true;
    }
};
````

# Quick Summary

| Situation | Result |
|---|---|
| Disconnected | Neither |
| 0 odd-degree vertices | Euler Circuit |
| 2 odd-degree vertices | Euler Path only |
| More than 2 odd-degree vertices | Neither |

## Euler vs Hamiltonian

- Euler: Every edge is used exactly once.
- Hamiltonian: Every vertex is visited exactly once.
