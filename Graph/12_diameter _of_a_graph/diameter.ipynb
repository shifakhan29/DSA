# Diameter of a Tree

## Problem

Given `n` nodes connected by `n - 1` undirected edges, find the length of the longest path in the tree. The path length is measured in the number of edges.

## Key Observation

The diameter can be found using two breadth-first searches:

1. Start from any node, such as node `0`, and find the farthest node.
2. Start from that farthest node and find the node farthest from it.

The distance obtained from the second BFS is the tree diameter.

## Algorithm

- Build an undirected adjacency list.
- Run BFS from node `0`.
- Store the farthest node from the first BFS.
- Run BFS again from that node.
- The maximum distance reached is the diameter.

A level-by-level BFS is used to calculate distance. Each BFS level represents one additional edge.

## Complexity

- **Time:** `O(n)`
- **Space:** `O(n)`

````cpp
#include <bits/stdc++.h>
using namespace std;

class Solution {
public:
    int treeDiameter(vector<vector<int>>& edges) {
        int n = edges.size() + 1;

        unordered_map<int, vector<int>> adj;

        // Build the undirected adjacency list.
        for (const auto& edge : edges) {
            int u = edge[0];
            int v = edge[1];

            adj[u].push_back(v);
            adj[v].push_back(u);
        }

        return findDiameter(adj, n);
    }

private:
    int findDiameter(unordered_map<int, vector<int>>& adj, int n) {
        // First BFS: find a node at the maximum distance from node 0.
        auto [farthestNode, firstDistance] = findFarthestNode(adj, n, 0);

        // Second BFS: find the distance from that node.
        auto [secondFarthestNode, diameter] =
            findFarthestNode(adj, n, farthestNode);

        return diameter;
    }

    pair<int, int> findFarthestNode(
        unordered_map<int, vector<int>>& adj,
        int n,
        int source
    ) {
        queue<int> q;
        vector<bool> visited(n, false);

        q.push(source);
        visited[source] = true;

        int maxDistance = 0;
        int farthestNode = source;

        while (!q.empty()) {
            int levelSize = q.size();

            for (int i = 0; i < levelSize; i++) {
                int current = q.front();
                q.pop();

                farthestNode = current;

                for (int neighbor : adj[current]) {
                    if (!visited[neighbor]) {
                        visited[neighbor] = true;
                        q.push(neighbor);
                    }
                }
            }

            if (!q.empty()) {
                maxDistance++;
            }
        }

        return {farthestNode, maxDistance};
    }
};
````

```text
edges = [[0,1], [1,2], [1,3], [3,4]]

        0

      2   3
          |
          4

First BFS from 0 finds node 2 or node 4.
Second BFS from node 4 reaches node 2 at distance 3.

Diameter = 3
```

## Edge Cases

- `edges = []`: one node, so the diameter is `0`.
- `edges = [[0,1]]`: two nodes, so the diameter is `1`.
- Chain: the diameter is the distance between the two end nodes.
- Star: the diameter is `2`.
