# Kosaraju's Algorithm

## 1. Definition

Kosaraju's algorithm counts the **Strongly Connected Components (SCCs)** of a directed graph.

A Strongly Connected Component is a maximal group of vertices such that every vertex is reachable from every other vertex within the same component.

## 2. Important Points

- Kosaraju's algorithm works only on directed graphs.
- The graph may be connected or disconnected.
- A vertex with no outgoing edge is still an SCC of size one.
- Reversing an edge changes `u -> v` into `v -> u`.

## 3. Why Reversing the Graph Is Required

The first DFS stores vertices in postorder finishing time. The second DFS processes vertices in reverse finishing order on the reversed graph. This ensures that every DFS call visits exactly one SCC.

## 4. Algorithm

1. Initialize a `visited` array and an empty stack.
2. Run DFS on the original graph for every unvisited vertex.
3. Push a vertex into the stack after visiting all of its neighbours.
4. Create a reversed adjacency list:

```cpp
adjReversed[v].push_back(u);
```

5. Reset `visited` to `false` for every vertex.
6. Pop vertices from the stack one by one.
7. Run DFS on the reversed graph for every unvisited vertex.
8. Increment `count` once for every DFS call.

## 5. Code

````python
class Solution {
public:
    void dfsFill(int u, vector<vector<int>>& adj,
                 vector<bool>& visited, stack<int>& st) {
        visited[u] = true;

        for (int &v : adj[u]) {
            if (!visited[v]) {
                dfsFill(v, adj, visited, st);
            }
        }

        st.push(u); // Push after DFS completion
    }

    void dfsTraverse(int u, vector<vector<int>>& adjReversed,
                     vector<bool>& visited) {
        visited[u] = true;

        for (int &v : adjReversed[u]) {
            if (!visited[v]) {
                dfsTraverse(v, adjReversed, visited);
            }
        }
    }

    int kosaraju(int V, vector<vector<int>>& adj) {
        stack<int> st;
        vector<bool> visited(V, false);

        // 1. First DFS: store DFS finishing order
        for (int i = 0; i < V; i++) {
            if (!visited[i]) {
                dfsFill(i, adj, visited, st);
            }
        }

        // 2. Reverse every directed edge
        vector<vector<int>> adjReversed(V);

        for (int u = 0; u < V; u++) {
            for (int &v : adj[u]) {
                adjReversed[v].push_back(u);
            }
        }

        // 3. Second DFS: process reverse finishing order
        int count = 0;
        visited = vector<bool>(V, false);

        while (!st.empty()) {
            int node = st.top();
            st.pop();

            if (!visited[node]) {
                dfsTraverse(node, adjReversed, visited);
                count++;
            }
        }

        return count;
    }
};
````

## 6. Dry Run

Consider:

```text
0 -> 1
1 -> 2
2 -> 0
1 -> 3
```

### Step 1: First DFS

- DFS from `0` visits `1`, `2`, and `3`.
- Vertices are pushed only after their DFS is complete.
- The stack therefore follows postorder finishing order.

### Step 2: Reverse Graph

```text
1 -> 0
2 -> 1
0 -> 2
3 -> 1
```

### Step 3: Second DFS

- Pop the top vertex from the stack.
- Run DFS on the reversed graph.
- Each traversal finds one SCC.
- `0`, `1`, and `2` form one SCC.
- `3` forms another SCC.
- The answer is `2`.

## 7. Complexity

Let `V` be the number of vertices and `E` the number of edges.

| Operation | Complexity |
|---|---|
| First DFS | O(V + E) |
| Reversing the graph | O(V + E) |
| Second DFS | O(V + E) |
| Total time | O(V + E) |
| Stack and visited array | O(V) |
| Reversed adjacency list | O(V + E) |

## 8. Important Remembering Points

- `DFS on original graph -> push in postorder`
- `Reverse all edges -> create adjReversed`
- `Reset visited -> start the second DFS`
- `One DFS call in reverse graph -> one SCC`
- `count -> number of SCCs`

## 9. Common Mistakes

- Writing `!visite[v]` instead of `!visited[v]`.
- Using `vector<int>& adj` instead of `vector<vector<int>>& adj` in `dfsFill`.
- Declaring `adjReversede` instead of `adjReversed`.
- Forgetting to reset `visited` before the second DFS.
- Counting stack entries instead of DFS traversals.
