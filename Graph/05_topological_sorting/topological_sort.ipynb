# Topological Sort (DFS)

## 1. Definition

Topological sorting is a linear ordering of vertices in a **Directed Acyclic Graph (DAG)** such that for every edge `u → v`, `u` appears before `v`.

**Applicable only to DAGs.**

## 2. Approach: DFS + Stack

* `visited[]` → Tracks visited nodes.
* `stack` → Stores nodes after completing DFS.
* Traverse every unvisited vertex.
* Visit all adjacent unvisited nodes recursively.
* Push the current node into the stack after exploring all neighbors.
* Pop the stack to get the topological order.

## 3. Code

```python
class Solution {
public:
    void DFS(vector<vector<int>>& adj,
             vector<bool>& visited, int u,
             stack<int>& s) {

        visited[u] = true;

        for (int &v : adj[u]) {
            if (!visited[v]) {
                DFS(adj, visited, v, s);
            }
        }

        s.push(u); // Push after DFS completion
    }

    vector<int> topoSort(vector<vector<int>>& adj) {
        int n = adj.size();

        vector<bool> visited(n, false);
        stack<int> s;

        for (int i = 0; i < n; i++) {
            if (!visited[i]) {
                DFS(adj, visited, i, s);
            }
        }

        vector<int> ans;

        while (!s.empty()) {
            ans.push_back(s.top());
            s.pop();
        }

        return ans;
    }
};
```

## 5. Complexity

| Complexity | Value    |
| ---------- | -------- |
| Time       | O(V + E) |
| Space      | O(V)     |

## 6. Important Points

* Topological sort is possible only for DAGs.
* Run DFS for every unvisited vertex (handles disconnected components).
* Push nodes after exploring all adjacent nodes.
* This implementation does not detect cycles.

**Remember:** `Visit → Explore → Push → Pop`

# Topological sort using BFS

```python
//BFS--> topological sort
class Solution {
  public:
    vector<int> topoSort(vector<vector<int>>& adj) {
    
        int n = adj.size();
        vector<int>indegree(n, 0 );
        queue<int>q;
        
        //1.store the indegree
        for(int u = 0 ; u< n ; u++){
            for(int &v : adj[u]){
                indegree[v]++;
            }
        }
        
        //2.fill the queue
        for(int u = 0 ; u<n ; u++){
            if(indegree[u] == 0)
            {
                q.push(u);
            }
        }
         //3.simple BFS
         vector<int>ans;
         while(!q.empty()){
             int u =q.front();
             q.pop();
             ans.push_back(u);//store in the ans
             
             for(int &v : adj[u]){
                 indegree[v]--;
                 if(indegree[v] ==0)
                 {
                     q.push(v);
                 }
             }
         }
        return ans;//TC: O(V+ E) || SC:O(N)
    }
};
```
