# Minimum Spanning Tree (MST)

## 1. What is a Spanning Tree?

A **Spanning Tree** is a subset of edges of a connected, undirected graph that:

- Connects all vertices.
- Contains no cycle.
- Has exactly **V - 1 edges**.

**Example:** If a graph has 5 vertices, its spanning tree must have exactly 4 edges.

## 2. What is a Minimum Spanning Tree (MST)?

A **Minimum Spanning Tree** is a spanning tree whose total edge weight is minimum among all possible spanning trees.

In simple words:

Connect all vertices with the **minimum possible total cost**, without forming a cycle.

### Important Properties

| Property | Explanation |
|---|---|
| Graph | Connected, undirected, weighted |
| Edges | Exactly V - 1 |
| Cycle | Not allowed |
| Objective | Minimum total edge weight |
| MST uniqueness | Not always unique |
| Negative weights | Allowed |

**Remember:**
- MST focuses on minimum total cost.
- Shortest Path focuses on minimum distance from a source.
- MST does NOT guarantee the shortest path between two vertices.

---

## 3. Algorithms to Find MST

There are two important algorithms:

1. **Prim's Algorithm**
2. **Kruskal's Algorithm**

Both find the Minimum Spanning Tree, but their approaches are different.
