# Kruskal's Algorithm

## What is it?
- Kruskal's Algorithm finds the Minimum Spanning Tree (MST) of a connected, weighted, undirected graph.
- It processes edges in ascending order of weight.
- It uses a greedy approach and Disjoint Set Union (DSU).
- An edge is accepted only if it joins two different components.
- An edge is rejected if it joins vertices already in the same component because it would create a cycle.

## Core Idea
Start with `V` separate components. Each vertex is initially its own component.

1. Sort all edges by weight.
2. Process edges from the smallest weight to the largest weight.
3. Find the representative/root of both endpoints using `find()`.
4. If the roots are different, use `Union()` and add the edge weight.
5. If the roots are equal, skip the edge to avoid a cycle.
6. Return the total selected edge weight.

An MST with `V` vertices contains exactly `V - 1` edges.

---

## Approach
1. Create `parent[V]` and `rank[V]`.
2. Initialize every vertex as its own parent and every rank as `0`.
3. Sort the edges using their third element, the weight.
4. Traverse each edge `{u, v, wt}`.
5. If `Union(parent, rank, u, v)` returns `true`, add `wt` to `sum`.
6. If it returns `false`, reject the edge because it creates a cycle.
7. Return `sum`.

---

## Why DSU is Used
`DSU` answers:
- Are two vertices in the same component?
- How do we merge two components?

For an edge `(u, v)`:
- If `find(u) != find(v)`, the vertices are in different components. Accept and union them.
- If `find(u) == find(v)`, they are already connected. Reject the edge.

---

## Core Logic

### 1. `find()` - Find the representative
```cpp
int find(int u, vector<int>& parent) {
    if (parent[u] == u)
        return u;

    return parent[u] = find(parent[u], parent); // Path compression
}
```
- If `parent[u] == u`, `u` is the root of its component.
- Otherwise, find the root recursively.
- Assigning `parent[u]` during the recursion performs path compression.

### 2. `Union()` - Merge two components
```cpp
int parent_u = find(u, parent);
int parent_v = find(v, parent);

if (parent_u == parent_v)
    return false;
```

- If both vertices have the same root, they are already connected.
- Returning `false` means this edge is rejected.

```cpp
if (rank[parent_u] > rank[parent_v]) {
    parent[parent_v] = parent_u;
}
else if (rank[parent_u] < rank[parent_v]) {
    parent[parent_u] = parent_v;
}
else {
    parent[parent_u] = parent_v;
    rank[parent_v]++;
}
```

- This is union by rank.
- Attach the component with smaller rank below the component with larger rank.
- If both ranks are equal, attach one tree and increase the selected root's rank.

### 3. Sorting edges
```cpp
auto lambda = [&](vector<int>& v1, vector<int>& v2) {
    return v1[2] < v2[2];
};

sort(edges.begin(), edges.end(), lambda);
```

- `v1[2]` and `v2[2]` contain the edge weights.
- The comparator sorts all edges from smallest to largest weight.

### 4. Accepting an edge
```cpp
if (Union(parent, rank, u, v)) {
    sum += wt;
}
```

- `Union()` returns `true` when it merges two different components.
- The edge weight is added to `sum`.
- `Union()` returns `false` for a cycle-forming edge.

## Dry Run
Consider:
```text
{0, 1, 10}, {0, 2, 6}, {0, 3, 5},
{1, 3, 15}, {2, 3, 4}
```

Initially:
```text
parent = [0, 1, 2, 3]
rank   = [0, 0, 0, 0]
sum = 0
```

Sorted edges:
```text
(2,3,4), (0,3,5), (0,2,6), (0,1,10), (1,3,15)
```

| Step | Edge | Roots | Action | Sum |
|---|---|---|---|---|
| 1 | `(2,3,4)` | `2 / 3` | Different components; select | 4 |
| 2 | `(0,3,5)` | `0 / 3` | Different components; select | 9 |
| 3 | `(0,2,6)` | `0 / 2` | Already connected; reject cycle | 9 |
| 4 | `(0,1,10)` | `0 / 1` | Different components; select | 19 |
| 5 | `(1,3,15)` | `0 / 0` | Already connected; reject cycle | 19 |

Selected edges:
```text
(2,3) = 4
(0,3) = 5
(0,1) = 10
```

**Minimum cost = 19**

## Algorithm
1. Initialize `sum = 0`.
2. Create `parent[V]` and `rank[V]`.
3. Initialize each vertex as its own parent and every rank with `0`.
4. Sort all edges by increasing weight.
5. For every edge `{u, v, wt}`:
   - Find the representative of `u`.
   - Find the representative of `v`.
   - If they are different, union them and add `wt` to `sum`.
   - Otherwise, skip the edge because it forms a cycle.
6. Return `sum`.

## Time Complexity
- Sorting `E` edges takes `O(E log E)`.
- With path compression and union by rank, DSU operations are nearly constant per operation.

**Time Complexity: `O(E log E)`**

## Space Complexity
- `parent[]`: `O(V)`
- `rank[]`: `O(V)`

**Auxiliary Space: `O(V)`**

---

## Critical Points
- The graph must be weighted, undirected, and connected for a complete MST.
- The edge format is `{u, v, weight}`.
- Sorting is greedy: select the smallest possible safe edge at every step.
- `find()` checks connectivity.
- `Union()` performs cycle detection and component merging.
- Path compression makes `find()` faster.
- Union by rank keeps the DSU trees shallow.
- Negative weights are allowed.
- For a disconnected graph, Kruskal's Algorithm finds a Minimum Spanning Forest.
- The `edges` vector is sorted in place.

## Remember
- **Kruskal = Sort edges + DSU**
- Accept an edge if the endpoints belong to different components.
- Reject an edge if both endpoints belong to the same component.
- Kruskal builds a forest and combines components until the MST remains.
- `find()` and `Union()` are the most important parts of this implementation.


````cpp
class Solution {
public:

    int find(int u, vector<int>& parent) {
        if (parent[u] == u) {
            return u;
        }

        return parent[u] = find(parent[u], parent);
    }

    bool Union(vector<int>& parent, vector<int>& rank, int u, int v) {

        int parent_u = find(u, parent);
        int parent_v = find(v, parent);

        if (parent_u == parent_v) {
            return false;
        }

        if (rank[parent_u] > rank[parent_v]) {
            parent[parent_v] = parent_u;
        }
        else if (rank[parent_u] < rank[parent_v]) {
            parent[parent_u] = parent_v;
        }
        else {
            parent[parent_u] = parent_v;
            rank[parent_v]++;
        }

        return true;
    }

    int kruskalsMST(int V, vector<vector<int>>& edges) {

        int sum = 0;

        vector<int> parent(V);
        vector<int> rank(V, 0);

        for (int i = 0; i < V; i++) {
            parent[i] = i;
        }

        auto lambda = [](const vector<int>& v1, const vector<int>& v2) {
            return v1[2] < v2[2];
        };

        sort(edges.begin(), edges.end(), lambda);

        int edgeCount = 0;

        for (auto& temp : edges) {

            int u = temp[0];
            int v = temp[1];
            int wt = temp[2];

            if (Union(parent, rank, u, v)) {
                sum += wt;
                edgeCount++;
            }

            if (edgeCount == V - 1) {
                break;
            }
        }

        return sum;
    }
};
````
