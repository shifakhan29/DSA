# Prim's Algorithm

## What is it?
- Prim's Algorithm finds the **Minimum Spanning Tree (MST)** of a connected, weighted, undirected graph.
- An MST connects all vertices without a cycle and has the minimum possible total edge weight.
- It starts with one node and repeatedly selects the minimum-weight edge that connects the visited set to an unvisited node.
- It uses a **greedy approach** and a **min-heap priority queue**.

## Approach
1. Create a `visited` array and mark every node as unvisited.
2. Create a min-heap storing `{edgeWeight, node}` pairs.
3. Start from node `0` by pushing `{0, 0}` into the heap.
4. Remove the minimum-weight pair from the heap.
5. Skip the node if it is already visited.
6. Mark the node as visited and add its weight to `sum`.
7. Traverse all unvisited neighbours and push `{neighbourWeight, neighbour}` into the heap.
8. Continue until the heap becomes empty.

## Algorithm
1. Initialize `visited[n] = false`.
2. Initialize `sum = 0`.
3. Create a min-heap: `priority_queue<pair<int, int>, vector<pair<int, int>>, greater<pair<int, int>>> pq`.
4. Push `{0, 0}` to represent the starting node.
5. Repeat while the heap is not empty:
   - Extract `{wt, node}` from the heap.
   - If `node` is already visited, skip it.
   - Mark `node` as visited.
   - Add `wt` to `sum`.
   - For every neighbouring node:
     - Push `{neighbour_wt, neighbour}` if it is not visited.
6. Return `sum` as the MST cost.

## Core Logic

```cpp
if (visited[node])
    continue;

visited[node] = true;
sum += wt;
```

- `visited[node]` ensures that every vertex is added to the MST only once.
- `sum += wt` adds the selected minimum edge weight to the MST cost.
- The minimum entry in the heap is always considered next.

```cpp
for (vector<int> &e : adj[node]) {
    int neighbour = e[0];
    int neighbour_wt = e[1];

    if (!visited[neighbour])
        pq.push({neighbour_wt, neighbour});
}
```

- Every unvisited neighbour becomes a candidate for the next MST edge.
- The heap automatically selects the smallest candidate edge.

## Time Complexity
- Every pushed heap entry takes `O(log V)` time.
- A node can have multiple heap entries, giving at most `O(E)` insertions.
- Visiting and scanning adjacency of every node takes `O(V + E)`.

**Time Complexity: `O((V + E) log V)`**

## Space Complexity
- `visited[]`: `O(V)`
- Priority queue: `O(E)` in the worst case
- Adjacency list: `O(V + E)`

**Extra algorithm space: `O(V + E)`**

## Critical Points
- Prim's Algorithm requires a **weighted graph**.
- The graph can be directed or undirected in representation, but Prim's MST is normally used for an **undirected, connected graph**.
- The priority queue stores `{weight, node}`, so `P` must represent weight first.
- `0` is used as the starting node in this code.
- The starting node contributes `0` weight to the MST sum.
- `visited[node]` prevents a node from being added more than once.
- The heap may contain duplicate or outdated entries; `visited` skips them.
- This implementation does not push visited neighbours, but because some entries were already pushed, the visited check is still required.
- If the graph is disconnected, this code returns the MST cost of only the connected component containing node `0`.
- For a disconnected graph, an MST is not possible for the entire graph, or we need to find an MST for every component.
- Negative weights are allowed by Prim's Algorithm because it works by selecting a minimum cut edge, not by relaxation based on source distances.

## Remember
- **Prim = Greedy + Min-Heap + Visited Set**
- Always grow the MST from the visited set to the unvisited set.
- Heap pair is `{edgeWeight, node}`.
- Minimum-cost edge among all edges crossing the cut is selected.
- The algorithm does not need to know whether an edge is part of the final MST before adding it.
- Prim's Algorithm builds one growing MST set, while Kruskal's Algorithm processes all edges in sorted order.

```cpp
class Solution {
	public:
	typedef pair<int, int>P;
	int primMST(vector<vector<vector<int>>> &adj) {
		int n = adj.size();
		vector<bool>visited(n, false);
		
		priority_queue<P, vector<P>, greater<P>> pq;
		int sum = 0;
		pq.push({0, 0}); // {wt,node}
		while (!pq.empty()) {
			auto curr = pq.top();
			int wt = curr.first;
			int node = curr.second;
			
			pq.pop();
			
			if (visited[node])
				continue;
			visited[node] = true;
			sum += wt;
			
			for (vector<int> &e : adj[node]) {
				int neighbour = e[0];
				int neighbour_wt = e[1];
				
				if (!visited[neighbour])
				
					pq.push({neighbour_wt, neighbour});
				
			}
		}
		
		return sum;
		
	}
};
```
