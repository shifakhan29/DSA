# Floyd Warshall Algorithm

Floyd-Warshall is an algorithm used to find the shortest paths between all pairs of vertices in a graph.

It is especially useful when we want to know the shortest path from every node to every other node, not just from one source node.


## 1. What does it do?

- It works on a weighted graph.
- It finds the shortest distance between every pair of vertices.
- This is called an All Pairs Shortest Path (APSP) problem.
- It is a dynamic programming approach.

### Real-life idea
Imagine a map where each city is a node and roads have distances. Floyd-Warshall tells us the shortest distance between every pair of cities in one run.


## 2. Key idea

We keep a matrix `dist` where:
- `dist[i][j]` = shortest known distance from node `i` to node `j`
- if there is no direct edge, we store a very large value like `INF`
- `dist[i][i] = 0` for every node

Then for every node `k` (the intermediate node), we check:

`dist[i][j] = min(dist[i][j], dist[i][k] + dist[k][j])`

This means:

- either the current shortest path from `i` to `j` is already better, or
- going through node `k` gives a shorter route.

So, every node is considered as a possible middle stop in the path.


## 3. Simple example

Suppose we have 4 vertices: 0, 1, 2, 3.

If we want to check whether going through node 2 gives a better path from 0 to 3, we look at:

`dist[0][3] = min(dist[0][3], dist[0][2] + dist[2][3])`

If this new value is smaller, we update it.

This happens for every pair of nodes and every possible intermediate node.


## 4. Step-by-step logic

1. Create a distance matrix with the direct edge weights.
2. Set `dist[i][i] = 0`.
3. For each `k` from 0 to `n-1`:
   - For each `i` from 0 to `n-1`:
      - For each `j` from 0 to `n-1`:
         - If `dist[i][k] + dist[k][j] < dist[i][j]`, update it.
4. At the end, the matrix contains all shortest paths.

This is called the triple loop because we use three nested loops.


## 5. Complexity

Let `V` be the number of vertices and `E` be the number of edges.

- Time Complexity: `O(V^3)`
- Space Complexity: `O(V^2)`

### Why is it `O(V^3)`?
Because we run three nested loops over all vertices:

`for k ... for i ... for j ...`

So the total work becomes roughly `V * V * V = V^3`.

This algorithm is best for graphs with a small or medium number of vertices.


## 6. When to use Floyd-Warshall?

Use this algorithm when:
- you need shortest paths between all pairs of nodes
- the graph is relatively small
- you want one matrix of shortest distances

Common use cases:
- network routing
- transportation maps
- graph analysis problems
- dynamic programming on graphs

It is not the best choice for huge graphs, because its time complexity is cubic.


## 7. Important points to remember

- Floyd-Warshall gives shortest paths between every pair of vertices.
- It is a dynamic programming algorithm.
- The main idea is to use every node as an intermediate step.
- It works with weighted graphs.
- It can also detect negative cycles if a value on the diagonal becomes negative.
- For graphs with no negative cycles, the algorithm is reliable.

### Quick formula
```
if dist[i][k] + dist[k][j] < dist[i][j]:
    dist[i][j] = dist[i][k] + dist[k][j]
```

This is the heart of Floyd-Warshall.


````python
class Solution {
public:
    void floydWarshall(vector<vector<int>>& dist) {
        int n = dist.size();
        const int INF = 1e8;

        for (int via = 0; via < n; via++) {
            for (int i = 0; i < n; i++) {
                for (int j = 0; j < n; j++) {
                    if (dist[i][via] != INF && dist[via][j] != INF) {
                        dist[i][j] = min(dist[i][j], dist[i][via] + dist[via][j]);
                    }
                }
            }
        }
    }
};

// Explanation:
// Here, via is the intermediate node.
// We try all possible middle nodes to find a shorter path from i to j.
// Time Complexity: O(n^3)
// Space Complexity: O(1) extra when using the matrix itself
````

## 8. Final summary

Floyd-Warshall is one of the easiest ways to understand shortest paths in a graph when we need answers for all pairs.

The main formula is simple:

`better path = dist[i][k] + dist[k][j]`

If this is smaller than the current value, we replace it.

That is the whole idea behind the algorithm.
