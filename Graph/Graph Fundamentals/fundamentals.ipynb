# Graph Fundamentals

## 1. What is a graph?

A **graph** is a way to describe things and the connections between them.

- **Vertex (node):** a thing, such as a city, person, or computer.
- **Edge:** a connection between two vertices, such as a road or friendship.

We write a graph as **G = (V, E)**, where `V` is the set of vertices and `E` is the set of edges.

**Example:** In a map, cities are vertices and roads are edges. In a social network, people are vertices and friendships are edges.

## 2. Important graph words

- **Adjacent neighbors:** vertices directly connected by an edge.
- **Degree:** number of edges touching a vertex. In directed graphs, use **in-degree** (incoming edges) and **out-degree** (outgoing edges).
- **Path:** a sequence of vertices connected by edges.
- **Cycle:** a path that returns to its starting vertex.
- **Connected graph:** every vertex can be reached from every other vertex. A disconnected graph has separate pieces.
- **Component:** one connected piece of an undirected graph.

## 3. Common graph types

- **Undirected:** edges work both ways, like a two-way road.
- **Directed (digraph):** edges have a direction, like `A -> B`.
- **Unweighted:** all edges are treated equally.
- **Weighted:** each edge has a cost, distance, or time.
- **Cyclic / acyclic:** whether the graph contains a cycle. A directed acyclic graph is called a **DAG**.

## 4. How to store a graph

For this example, the edges are `0-1`, `0-2`, `1-3`, and `2-3`.

### Adjacency list (usually the best default)

For each vertex, store its neighbors. It uses **O(V + E)** space and is efficient for checking a vertex's neighbors. In an undirected graph, store each edge at both endpoints.

### Adjacency matrix

Use a `V x V` table. `matrix[u][v]` indicates whether an edge exists (or stores its weight). It uses **O(V^2)** space, but checking whether a particular edge exists takes **O(1)** time.

### Edge list

Store all edges as pairs (or triples for weighted edges), for example `[(0, 1), (0, 2)]`. It is simple and useful when processing all edges.

### C++ example: build an undirected adjacency list

Each edge is added in both directions. For a directed graph, add only `v` to `adj[u]`.

```cpp
#include <iostream>
#include <vector>
using namespace std;

int main() {
    int V = 4;
    vector<vector<int>> adj(V);
    vector<pair<int, int>> edges = {{0, 1}, {0, 2}, {1, 3}, {2, 3}};

    for (auto [u, v] : edges) {
        adj[u].push_back(v);
        adj[v].push_back(u);
    }

    for (int u = 0; u < V; ++u) {
        cout << u << ": ";
        for (int v : adj[u]) cout << v << " ";
        cout << "\n";
    }
}
```

## 5. Traversing a graph: BFS and DFS

A traversal visits vertices while avoiding repeated visits. Always track visited vertices: graphs can contain cycles.

### Breadth-first search (BFS)

BFS explores **nearby vertices first**, like ripples spreading across water. It uses a **queue** (first in, first out). In an unweighted graph, BFS finds a shortest path measured by number of edges.

### Depth-first search (DFS)

DFS follows one route as far as it can, then **backtracks**. It uses recursion (the call stack) or an explicit stack. DFS is useful for reachability, components, and cycle-related problems.

For an adjacency-list graph, either traversal takes **O(V + E)** time and up to **O(V)** extra space.

````python
//traversal --> BFS
void bfs(vector<vector<int>>& adj, int u, vector<bool>& visited , vector<int>&ans){
    queue<int>q;
    q.push(u);
    visited[u] = true;
    ans.push_back(u);//push the u in ans

    while(!q.empty()){
        int u = q.front();
        q.pop();
        for(int &v : adj[u]){
            if(!visited[v])
            {
                q.push(v);
                visited[v] = true;
                ans.push_back(v);//push the ans
            }
        }
    }

}
vector<int>BFS(vector<vector<int>>&adj, int V){
    vector<bool>visited(V,false);
    vector<int>ans;
    for(int i=0; i<V ; i++){
        if(!visited[i]){
            bfs(adj, i, visited, ans);
        }
    }
    return ans;//TC : O(V + E)  || SC: O(V)
}
````

````python
// Traversal --> DFS

void dfs(vector<vector<int>>& adj, int u,
         vector<bool>& visited, vector<int>& ans) {

    if (visited[u])
        return;

    visited[u] = true;
    ans.push_back(u);

    for (int &v : adj[u]) {
        if (!visited[v]) {
            dfs(adj, v, visited, ans);
        }
    }
}

vector<int> DFS(vector<vector<int>>& adj, int V) {

    vector<bool> visited(V, false);
    vector<int> ans;

    for (int i = 0; i < V; i++) {
        if (!visited[i]) {
            dfs(adj, i, visited, ans);
        }
    }

    return ans; //TC : O(V + E)  || SC: O(V)
}
````

## 6. Important details and common mistakes

- A traversal starting at one vertex only visits its reachable component. To visit a disconnected graph, loop over every vertex and start a traversal from each unvisited one.
- In a directed graph, add only the direction given by each edge. For `u -> v`, add `v` to `adj[u]`; do not automatically add `u` to `adj[v]`.
- Mark a neighbor visited **when adding it to the queue/stack**, not later. This avoids adding it many times.
- BFS gives shortest paths only when every edge has the same cost (usually an unweighted graph). For weighted shortest paths, learn Dijkstra's algorithm or other weighted-graph algorithms.
- The order printed by BFS/DFS can change if the neighbors are stored in a different order. The reachable set stays the same.

## 7. Quick complexity guide

| Representation / operation | Time or space |
|---|---:|
| Adjacency list space | O(V + E) |
| Adjacency matrix space | O(V^2) |
| Check edge in adjacency list | O(degree of vertex) |
| Check edge in adjacency matrix | O(1) |
| BFS or DFS with adjacency list | O(V + E) |

`V` = number of vertices, `E` = number of edges. For an undirected graph, each edge is usually stored twice in an adjacency list.

## 8. How to approach a graph problem

1. Identify what the vertices represent.
2. Identify what makes two vertices connected; check whether edges are directed or weighted.
3. Choose a representation, usually an adjacency list.
4. Decide what the question needs: reachability, shortest unweighted path, components, cycle detection, or ordering.
5. Choose the algorithm (often BFS or DFS), track visited vertices, and account for disconnected components.

**Memory trick:** BFS = queue = layers / nearest first. DFS = stack or recursion = go deep, then backtrack.
