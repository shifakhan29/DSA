# 0-1 BFS (Breadth-First Search)

## 1. Motivation
- **Dijkstra's Algorithm** solves single-source shortest path problems with non-negative edge weights.
- **0-1 BFS** is an optimized version of Dijkstra's for graphs where edge weights are restricted to **0 or 1**.
- It provides better time complexity by eliminating the need for a priority queue.

## 2. Core Concepts

- **Data Structure:** Uses a `Deque` (Double-Ended Queue).
- **Weight = 0:** Push the adjacent node to the **front** of the deque.
- **Weight = 1:** Push the adjacent node to the **back** of the deque.

### Why Deque?
The deque maintains nodes in non-decreasing order of their tentative distances, allowing us to process closer nodes first without a priority queue.

## 3. Algorithm

1. Initialize `dist[]` with infinity.
2. Set the source distance to `0`.
3. Push the source node into the deque.
4. While the deque is not empty:
   - Extract the front node.
   - Visit its adjacent nodes.
   - Calculate the new distance:

   `newDist = dist[u] + weight`

   - If `newDist < dist[v]`:
     - Update `dist[v]`.
     - If weight is `0`, push `v` to the **front**.
     - If weight is `1`, push `v` to the **back**.
5. Return the shortest distances.

## 4. Core Logic

| Edge Weight | Operation | Reason |
|---|---|---|
| 0 | `push_front()` | Same distance; process first |
| 1 | `push_back()` | Distance increases by 1 |

**Remember:**
- Weight `0` → Front
- Weight `1` → Back

## 5. Complexity Analysis

Let:
- `V` = Number of vertices
- `E` = Number of edges

| Complexity | Value |
|---|---|
| Time | `O(V + E)` |
| Space | `O(V)` |

## 6. Dijkstra vs 0-1 BFS

| Feature | Dijkstra | 0-1 BFS |
|---|---|---|
| Edge weights | Non-negative | Only 0 or 1 |
| Data structure | Min-Heap / Priority Queue | Deque |
| Time complexity | `O((V + E) log V)` | `O(V + E)` |
| Use case | General weighted graphs | Binary-weighted graphs |

## 7. Applications

- Shortest path problems with edge weights `0` and `1`.
- Grid-based shortest path problems.
- Problems where moving between cells has a cost of either `0` or `1`.


## Quick Revision

> **0-1 BFS = BFS + Deque + Binary Edge Weights**

- `0` weight → Front
- `1` weight → Back
- Shortest path in `O(V + E)`.
- Specialized optimization of Dijkstra.
