# Bellman-Ford Algorithm

## 1. Overview
- Bellman-Ford is a **Single Source Shortest Path (SSSP)** algorithm.
- It finds the shortest distance from a given source vertex to all other vertices in a weighted graph.
- It works with both **positive and negative edge weights**.
- It can detect **negative weight cycles**.

## 2. Key Concepts

### 1. Handling Negative Weights
- Unlike Dijkstra's algorithm, Bellman-Ford supports negative edge weights.
- Dijkstra does not work reliably with negative edge weights because its greedy approach assumes that once a node is processed, its shortest distance is finalized.

### 2. Relaxation
Relaxation means updating the distance of a vertex if a shorter path is found.

**Formula:**

dist[u] + wt < dist[v]

If the condition is true:

dist[v] = dist[u] + wt

Where:
- `u` = Source vertex of an edge.
- `v` = Destination vertex.
- `wt` = Edge weight.
- `dist[]` = Shortest known distance from the source.

### 3. Number of Relaxations
- Relax all edges **V - 1 times**, where V is the number of vertices.
- Why V - 1?
  - A shortest path without a negative cycle can contain at most V - 1 edges.
  - Each iteration considers paths using one additional edge.

### 4. Negative Cycle Detection
A negative cycle is a cycle whose total edge weight is negative.

**Detection Process:**
1. First, relax all edges V - 1 times.
2. Perform one additional iteration (Vth iteration).
3. If any distance can still be reduced, a negative weight cycle exists.

**Condition:**

`dist[u] + wt < dist[v]`

If this condition is still true during the Vth iteration, a reachable negative cycle exists.

> Note: This detects negative cycles reachable from the source vertex.

### 5. Directed and Undirected Graphs
- Bellman-Ford is commonly implemented using directed edges.
- For an undirected graph, represent each edge as two directed edges:
  - `u -> v`
  - `v -> u`

**Important:** An undirected graph containing a negative-weight edge creates a negative-weight cycle because the edge can be traversed in both directions.

---

## 3. Algorithm

1. Initialize the distance array with `INT_MAX`.
2. Set the source vertex distance to `0`.
3. Repeat V - 1 times:
   - Traverse all edges.
   - If the source vertex is reachable and relaxation is possible, update the destination distance.
4. Traverse all edges one more time.
5. If any distance can still be reduced, report a negative cycle.
6. Otherwise, return the shortest distances.

---

## 4. Complexity Analysis

Let:
- V = Number of vertices.
- E = Number of edges.

| Complexity | Value | Explanation |
|---|---|---|
| Time | O(V * E) | Relaxing all E edges V - 1 times |
| Space | O(V) | Distance array |
| Negative Cycle Detection | O(E) | One additional traversal |

**Overall Time Complexity:**

`O(V * E)`

**Overall Auxiliary Space Complexity:**

`O(V)`

---

## 5. Bellman-Ford vs Dijkstra

| Feature | Bellman-Ford | Dijkstra |
|---|---|---|
| Negative weights | Yes | No |
| Negative cycle detection | Yes | No |
| Approach | Edge relaxation | Greedy |
| Time Complexity | O(V * E) | O((V + E) log V) with a min-heap |
| Shortest path | Single source | Single source |

---

## 6. Remember

- **V - 1 iterations** for shortest paths.
- **Vth iteration** for negative cycle detection.
- Supports negative edge weights.
- Detects only negative cycles reachable from the source.
- Uses edge relaxation.
- Time Complexity: `O(V * E)`.
- Auxiliary Space Complexity: `O(V)`.
- For undirected graphs, add edges in both directions. A negative-weight undirected edge creates a negative cycle.

### Quick Revision

1. Initialize `dist[] = INT_MAX`.
2. Set `dist[source] = 0`.
3. Relax all edges V - 1 times.
4. Relax all edges one more time.
5. If distance decreases -> Negative cycle exists.
6. Otherwise -> Shortest distances obtained.

# Code
```cpp
class Solution {
	public:
	vector<int> bellmanFord(int V, vector<vector<int>> & edges, int src) {
		vector<int>distance(V, INT_MAX);
		distance[src] = 0;
		for (int i = 0 ; i < V - 1 ; i++) {
			for (auto &e : edges) {
				int u = e[0];
				int v = e[1];
				int wt = e[2];
				if (distance[u] != INT_MAX && distance[u] + wt < distance[v]) {
					distance[v] = distance[u] + wt;
				}
			}
			
		}
		for (auto &edge : edges) {
			int u = edge[0];
			int v = edge[1];
			int wt = edge[2];
			if (distance[u] != INT_MAX && distance[u] + wt < distance[v]) {
				return {-1};
			}
		}
		return distance;
		//TC:O(V*E) || SC: O(V*E)
	}
};
```
