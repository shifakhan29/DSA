# Multi-Source BFS (Breadth-First Search)

## 1. What is BFS?

**Breadth-First Search (BFS)** is a graph traversal algorithm that explores nodes level by level, starting from a source node.

- First, it visits the source node.
- Then, it visits all its immediate neighbours.
- Next, it explores the neighbours of those nodes.
- It continues until all reachable nodes are visited.

### Key Idea
BFS uses a **Queue (FIFO)** to process nodes in the order they are discovered.

For an unweighted graph, BFS guarantees the shortest path (minimum number of edges) from the source to every reachable node.

---

## 2. What is Multi-Source BFS?

**Multi-Source BFS** is an extension of normal BFS in which traversal starts from multiple source nodes simultaneously.

Instead of inserting one source into the queue, we insert all source nodes initially.

### How It Works

1. Identify all the source nodes.
2. Mark each source as visited.
3. Insert all sources into the queue.
4. Process nodes level by level.
5. Visit unvisited neighbours and calculate their distances.

### Key Idea

Multi-Source BFS calculates the minimum distance of every reachable node from its **nearest source**.

Mathematically:

\[
dist[v] = \min_{s \in S} dist(s,v)
\]

Where:
- \(S\) = Set of all source nodes.
- \(dist(s,v)\) = Shortest distance from source `s` to node `v`.

---

## 3. Application in 2D Grids

Multi-Source BFS is particularly useful for grid-based problems where multiple cells act as starting points.

### Common Applications

- Rotting Oranges (LeetCode 994).
- Distance from Nearest Zero (LeetCode 542).
- Walls and Gates (LeetCode 286).
- Finding minimum time to spread something across a grid.

### Example: Rotting Oranges

Consider the following grid:

```text
2 1 1
1 1 0
0 1 1
```

Here:
- `2` represents a rotten orange (source).
- `1` represents a fresh orange.
- `0` represents an empty cell.

### Approach

1. Traverse the entire grid.
2. Insert all rotten oranges into the queue.
3. Count the total number of fresh oranges.
4. Process the queue level by level.
5. Whenever a rotten orange reaches an adjacent fresh orange:
   - Mark it as rotten.
   - Decrease the fresh-orange count.
   - Insert its position into the queue.
6. Each BFS level represents one minute.
7. If fresh oranges remain after BFS, return `-1`; otherwise, return the elapsed time.

**Why Multi-Source BFS?**

Because all initially rotten oranges spread simultaneously. Starting BFS from only one rotten orange would not correctly represent the simultaneous spreading process.

## 4. BFS vs DFS

| Feature | BFS | DFS |
|---|---|---|
| Traversal | Level by level | Depth-wise |
| Data Structure | Queue | Stack / Recursion |
| Shortest Path | Guaranteed in unweighted graphs | Not guaranteed |
| Multi-Source | Naturally supported | Possible, but not ideal for shortest distances |
| Grid Problems | Minimum distance/time | Connectivity and path exploration |

### Why Not Use DFS for Shortest-Path Problems?

DFS explores one path deeply before exploring other paths.

Consider a graph where a node can be reached through two different paths:

- Path 1: 2 edges.
- Path 2: 5 edges.

DFS might explore the longer path first.

If we mark the node as visited immediately, we may prevent exploring a shorter path later.

BFS avoids this problem because it explores nodes in increasing order of distance.

Therefore, the first time BFS visits a node, it has found the shortest distance to that node in an unweighted graph.

## 5. Complexity Analysis

Let:
- `V` = Number of vertices.
- `E` = Number of edges.
- `R` = Number of rows.
- `C` = Number of columns.

### Graph

| Complexity | Value |
|---|---|
| Time | O(V + E) |
| Space | O(V) |

### 2D Grid

| Complexity | Value |
|---|---|
| Time | O(R × C) |
| Space | O(R × C) |

Each cell is processed at most once, and each cell has at most four neighbours in a standard 2D grid.

---

## 6. Remember

- BFS uses a **Queue (FIFO)**.
- Normal BFS starts from one source.
- Multi-Source BFS starts from multiple sources simultaneously.
- Insert all sources into the queue before starting traversal.
- BFS finds the shortest distance in unweighted graphs.
- Multi-Source BFS finds the nearest-source distance.
- In grid problems, one BFS level often represents one unit of time.
- DFS is suitable for connectivity and exploration but does not guarantee shortest paths.

### Core Logic

**Single-Source BFS:**
```cpp
q.push(source);
```

**Multi-Source BFS:**
```cpp
for (int i = 0; i < n; i++) {
    if (isSource[i]) {
        q.push(i);
        visited[i] = true;
    }
}
```

**Difference:**

> Single-Source BFS → Shortest distance from one source.

> Multi-Source BFS → Shortest distance from the nearest source.
