```python
class Solution {
public:

    pair<int, int> BFS(unordered_map<int, vector<int>>& adj,
                       int n, int source) {

        queue<int> q;
        vector<bool> visited(n, false);

        q.push(source);
        visited[source] = true;

        int maxdistance = 0;
        int farthest = source;

        while(!q.empty()) {

            int s = q.size();

            for(int i = 0; i < s; i++) {

                int curr = q.front();
                q.pop();

                farthest = curr;

                for(int& neigh : adj[curr]) {

                    if(!visited[neigh]) {
                        visited[neigh] = true;
                        q.push(neigh);
                    }
                }
            }

            if(!q.empty())
                maxdistance++;
        }

        return {farthest, maxdistance};
    }

    int findDia(unordered_map<int, vector<int>>& adj, int n) {

        auto [farthestPoint, dist1] = BFS(adj, n, 0);

        auto [otherPoint, diameter] = BFS(adj, n, farthestPoint);

        return diameter;
    }

    int treeDiameter(vector<vector<int>>& graph1,
                     vector<vector<int>>& graph2) {

        int n1 = graph1.size() + 1;
        int n2 = graph2.size() + 1;

        unordered_map<int, vector<int>> adj1, adj2;

        for(auto& e : graph1) {
            int u = e[0];
            int v = e[1];

            adj1[u].push_back(v);
            adj1[v].push_back(u);
        }

        for(auto& e : graph2) {
            int u = e[0];
            int v = e[1];

            adj2[u].push_back(v);
            adj2[v].push_back(u);
        }

        int d1 = findDia(adj1, n1);
        int d2 = findDia(adj2, n2);

        int radius1 = (d1 + 1) / 2;
        int radius2 = (d2 + 1) / 2;

        return max({d1, d2, radius1 + radius2 + 1});
    }
};
```
