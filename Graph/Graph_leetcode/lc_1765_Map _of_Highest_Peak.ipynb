# Leetcode 1765 : Mapping Of High Peak 

## Approach

1. Create a `height` matrix and initialize every cell with `-1`.
2. Traverse `isWater` and find all water cells.
3. Set the height of every water cell to `0` and push it into the queue.
4. Perform BFS from all water cells simultaneously.
5. Remove a cell from the queue and check its four neighbors.
6. If a neighbor is inside the grid and is unvisited (`height == -1`), set:

   ```cpp
   height[X][Y] = height[x][y] + 1;
   ```

7. Push the neighbor into the queue and continue the BFS.
8. Return the `height` matrix.

## Algorithm: Multiple-Source BFS

Treat every water cell as a separate BFS source.

### Core Logic

1. Initialize `height[n][m]` with `-1`. A value of `-1` means that the land cell is not visited yet.
2. Traverse `isWater` and find every water cell.
3. Set each water cell's height to `0` and push it into the queue.
4. Perform BFS starting from all water cells simultaneously.
5. For the current cell, check all four neighbors: up, down, left, and right.
6. If a neighbor is inside the grid and has not been visited, set:

   ```cpp
   height[X][Y] = height[x][y] + 1;
   ```

7. Push the newly assigned neighbor into the queue so it can expand the distance further.

### Concept Used

**Multiple-source BFS** starts with every water cell in the queue at distance `0`. The BFS then spreads outward level by level. Therefore, every land cell receives the shortest distance from the nearest water cell.

The `height` matrix acts as both:

- the visited matrix, and
- the final height matrix.

The condition `height[X][Y] == -1` ensures that each land cell is assigned only once, on the first and shortest BFS visit.

### Example

```text
isWater:
[1, 0, 0]
[0, 0, 0]
[0, 0, 1]

height:
[0, 1, 2]
[1, 2, 1]
[2, 1, 0]
```

### Complexity

- **Time Complexity:** `O(n * m)` — each cell is enqueued and processed once.
- **Space Complexity:** `O(n * m)` — for the height matrix and queue.

````python
class Solution {
public:
    typedef pair<int, int> P;
    vector<vector<int>> directions = {{1, 0}, {0, 1}, {-1, 0}, {0, -1}};
    vector<vector<int>> highestPeak(vector<vector<int>>& isWater) {
        int n = isWater.size();
        int m = isWater[0].size();
        queue<P> q;

        vector<vector<int>> height(n, vector<int>(m,-1));
        for (int i = 0; i < n; i++) {
            for (int j = 0; j < m; j++) {
                if (isWater[i][j] == 1) {
                    height[i][j] = 0;
                    q.push({i, j});
                }
            }
        }
        while (!q.empty()) {
            int s = q.size();
            while (s--) {
                auto curr = q.front();
                q.pop();

                int x = curr.first;
                int y = curr.second;

                for (auto& dir : directions) {
                    int X = x + dir[0];
                    int Y = y + dir[1];
                    if (X >= 0 && Y >= 0 && X <= n - 1 && Y <= m - 1 &&
                        height[X][Y] == -1) {
                        height[X][Y] = height[x][y] + 1;
                        q.push({X, Y});
                    }
                }
            }
        }
        return height;
    }//TC: O(m*n) || SC: O(m*n)
};
````
