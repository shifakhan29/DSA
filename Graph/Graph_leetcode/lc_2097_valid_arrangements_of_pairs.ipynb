# LeetCode 2097: Valid Arrangement of Pairs

## Topic
Directed graph + Eulerian path + Hierholzer's algorithm.

## Problem
Given `pairs[i] = [starti, endi]`, arrange all pairs so that the second value of one pair becomes the first value of the next pair.

In graph terms:
- Each pair `[u, v]` is a directed edge `u -> v`.
- We need an Eulerian path that uses every edge exactly once.
- The answer is guaranteed to exist.

## Eulerian Path Condition
- A directed graph has an Eulerian path if all edges belong to one connected component.
- Count the indegree and outdegree of every vertex.
- Usually, one vertex has `outdegree - indegree = 1`; that vertex is the starting point.
- If no such vertex exists, the graph is Eulerian/cyclic, so any vertex with an outgoing edge can be the start.

## Approach: Hierholzer's Algorithm
1. Build an adjacency list containing directed edges.
2. Count indegree and outdegree for every vertex.
3. Find a starting vertex:
   - If a vertex has `outdegree - indegree == 1`, choose it.
   - Otherwise, choose the first starting point of a pair.
4. Use an iterative stack starting at this vertex.
5. While the current vertex has a remaining neighbour:
   - Remove one neighbour from the adjacency list.
   - Push that neighbour onto the stack.
6. If no neighbour remains:
   - Backtrack by adding the current vertex to `path`.
   - Pop it from the stack.
7. Reverse `path` because vertices are recorded during backtracking.
8. Convert consecutive vertices in `path` into pairs.

## Dry Run
Input: `[[5,1], [4,5], [5,6], [1,4]]`

Edges:
- `5 -> 1`
- `4 -> 5`
- `5 -> 6`
- `1 -> 4`

Degrees:
- Vertex `5`: outdegree `2`, indegree `1` -> unique Euler path start
- Vertex `6`: indegree `1`, outdegree `0` -> Euler path end

The algorithm starts at `5` and can produce a path like:
`5 -> 1 -> 4 -> 5 -> 6`

Converting it into answer pairs gives:
`[[5,1], [1,4], [4,5], [5,6]]`

## Why Reverse the Path?
Hierholzer's algorithm pushes vertices while moving forward and records vertices while backtracking.
The recorded order is reverse, so reversing it gives the actual Eulerian path.

## Important Code Details
- Use `p[0]` for `u` and `p[1]` for `v`.
- Do not use `pairs[0]` for every pair; that would always read the first pair.
- Use `stack<int>` or `vector<int>` as the stack. `stack<int>` uses `push()` and `pop()`, not `push_back()`.
- Popping an adjacent vertex from `adj[u]` makes each directed edge available exactly once.

## Complexity
- Time: `O(V + E)`
- Space: `O(V + E)`

Here, `V` is the number of distinct vertices and `E` is `pairs.size()`.

```python
class Solution {
public:
    vector<vector<int>> validArrangement(vector<vector<int>>& pairs) {
        unordered_map<int, vector<int>> adj;
        unordered_map<int, int> indegree, outdegree;

        // 1. Build the directed graph.
        for (auto& p : pairs) {
            int u = p[0];
            int v = p[1];

            adj[u].push_back(v);
            indegree[v]++;
            outdegree[u]++;
        }

        // 2. Choose the Eulerian path start.
        int startNode = pairs[0][0];
        for (auto& p : pairs) {
            int node = p[0];
            if (outdegree[node] - indegree[node] == 1) {
                startNode = node;
                break;
            }
        }

        // 3. Hierholzer's algorithm using an iterative stack.
        vector<int> path;
        stack<int> st;
        st.push(startNode);

        while (!st.empty()) {
            int node = st.top();

            if (!adj[node].empty()) {
                int neighbour = adj[node].back();
                adj[node].pop_back();
                st.push(neighbour);
            } else {
                path.push_back(node);
                st.pop();
            }
        }

        reverse(path.begin(), path.end());

        // 4. Convert the vertex path into edge pairs.
        vector<vector<int>> ans;
        for (int i = 0; i + 1 < (int)path.size(); i++) {
            ans.push_back({path[i], path[i + 1]});
        }

        return ans; // Time: O(V + E), Space: O(V + E)
    }
};
```
