# LeetCode 1584: Min Cost to Connect All Points

## Problem

Given `n` points in a 2D plane, connect all points so that the total cost is minimum.

- The cost of an edge between two points is their **Manhattan distance**:

```text
|x1 - x2| + |y1 - y2|
```

- We can connect any pair of points directly.
- We need a Minimum Spanning Tree (MST) containing all `n` points.

## Key Observation

The complete graph has an edge between every pair of points. The edge weight is the Manhattan distance. Finding the cheapest way to connect all points is equivalent to finding the MST of this complete graph.

## Approach: Prim's Algorithm

Start with one point and repeatedly choose the minimum-cost edge that connects the visited set to an unvisited point.

1. Build an adjacency list containing every pair of points.
2. Use a min-heap storing `{edgeWeight, node}`.
3. Start from node `0` with weight `0`.
4. Pop the smallest available edge.
5. Skip the node if it is already visited.
6. Mark it visited and add its weight to `sum`.
7. Push all unvisited neighbours and their edge weights into the heap.
8. When the heap becomes empty, `sum` is the minimum total cost.

## Why Prim's Algorithm Works

- The graph is connected because every pair of points has an edge.
- Each step chooses the minimum edge crossing the cut between visited and unvisited points.
- This greedy choice produces an MST by the cut property.
- A visited node is never added twice, so no cycle is included.

## Algorithm

1. Initialize `visited[n] = false`.
2. Build the complete undirected graph:
   - For every pair `(i, j)`, calculate the Manhattan distance.
   - Add `{dist, j}` to `adj[i]`.
   - Add `{dist, i}` to `adj[j]`.
3. Push `{0, 0}` into a min-heap.
4. Pop `{wt, node}` repeatedly:
   - If `visited[node]` is true, continue.
   - Mark `node` visited and add `wt` to `sum`.
   - Push every unvisited neighbour into the heap.
5. Return `sum`.

## Core Logic

```cpp
int dist = abs(x2 - x1) + abs(y2 - y1);
adj[i].push_back({dist, j});
adj[j].push_back({dist, i});
```

- Build all undirected edges once.
- `dist` is the Manhattan distance.
- Both adjacency entries represent the same edge.

```cpp
if (visited[node])
    continue;

visited[node] = true;
sum += wt;
```

- Skip stale heap entries whose node was already selected.
- Mark the selected node as visited.
- Add the edge weight used to reach this node.
- The weight of the starting node is `0`, so it is visited without extra cost.

```cpp
if (!visited[neighbour]) {
    pq.push({neighbour_wt, neighbour});
}
```

Only unvisited nodes are pushed as candidates. The min-heap ensures that the smallest current candidate is processed first.

## Complexity Analysis

- Building the graph: `O(n^2)`
- Prim's traversal with a heap: `O(n^2 log n)`
- Overall: `O(n^2 log n)`
- Space: `O(n^2)` for the adjacency list

A more efficient Prim implementation can use Manhattan distances on the fly and run in `O(n^2)` time with `O(n)` extra space.

## Important Points

- The graph is complete, so every pair is a valid edge.
- `pair<int, int>` stores `{weight, node}` because heap ordering should prioritize weight first.
- `greater<pair<int, int>>` makes `priority_queue` behave as a min-heap.
- Stale heap entries must be skipped using `visited[node]`.
- Starting from any node produces the same minimum cost.
- Inserting both directions is required because the graph is undirected.

## Edge Cases

- `n = 1`: No edges are built, `{0, 0}` is processed, and the result is `0`.
- Duplicate coordinates: The distance is `0`, so one point can be added for free.
- Negative coordinates: `abs()` correctly computes the Manhattan distance.

## Remember

- **Problem type:** Minimum Spanning Tree.
- **Edge weight:** Manhattan distance.
- **Algorithm:** Prim's Algorithm.
- **Heap pair:** `{edgeWeight, node}`.
- **Result:** Minimum total edge weight that connects all points.
- **Greedy rule:** Always choose the minimum edge between the visited set and unvisited nodes.

````cpp
class Solution {
public:
    typedef pair<int, int> P; // {wt, node}

    int minCostConnectPoints(vector<vector<int>>& points) {
        int n = points.size();
        vector<bool> visited(n, false);
        vector<vector<P>> adj(n);

        for (int i = 0; i < n; i++) {
            for (int j = i + 1; j < n; j++) {
                int x1 = points[i][0];
                int y1 = points[i][1];
                int x2 = points[j][0];
                int y2 = points[j][1];

                int dist = abs(x2 - x1) + abs(y2 - y1);
                adj[i].push_back({dist, j});
                adj[j].push_back({dist, i});
            }
        }

        int sum = 0;
        priority_queue<P, vector<P>, greater<P>> pq;
        pq.push({0, 0});

        while (!pq.empty()) {
            auto curr = pq.top();
            pq.pop();

            int wt = curr.first;
            int node = curr.second;

            if (visited[node])
                continue;

            visited[node] = true;
            sum += wt;

            for (auto& e : adj[node]) {
                int neighbour_wt = e.first;
                int neighbour = e.second;

                if (!visited[neighbour]) {
                    pq.push({neighbour_wt, neighbour});
                }
            }
        }

        return sum;
    }
};
````

## Approach: Kruskal's Algorithm

Sort all possible point-to-point edges by Manhattan distance, then use Disjoint Set Union (DSU) to accept only edges that join two different components.

1. Create every undirected edge between a pair of points.
2. Sort the edges in increasing order of weight.
3. Initialize each point as its own DSU component.
4. For each edge `{u, v, weight}`:
   - If `find(u)` equals `find(v)`, skip it to avoid a cycle.
   - Otherwise, unite the components and add the weight to the answer.
   - Stop after accepting `n - 1` edges, as that forms an MST.

### Complexity

- Generating edges: `O(n^2)`
- Sorting edges: `O(n^2 log n)`
- DSU operations: almost `O(1)` amortized with path compression and union by rank
- Overall: `O(n^2 log n)` time and `O(n^2)` edge storage

````cpp
class Solution {
public:

    int find(int u, vector<int>& parent) {
        if (parent[u] == u) {
            return u;
        }

        return parent[u] = find(parent[u], parent);
    }

    bool Union(vector<int>& parent, vector<int>& rank, int u, int v) {

        int parent_u = find(u, parent);
        int parent_v = find(v, parent);

        if (parent_u == parent_v) {
            return false;
        }

        if (rank[parent_u] > rank[parent_v]) {
            parent[parent_v] = parent_u;
        }
        else if (rank[parent_u] < rank[parent_v]) {
            parent[parent_u] = parent_v;
        }
        else {
            parent[parent_u] = parent_v;
            rank[parent_v]++;
        }

        return true;
    }

    int minCostConnectPoints(vector<vector<int>>& points) {

        int n = points.size();

        vector<vector<int>> edges;

        // Generate all possible edges
        for (int i = 0; i < n; i++) {
            for (int j = i + 1; j < n; j++) {

                int x1 = points[i][0];
                int y1 = points[i][1];

                int x2 = points[j][0];
                int y2 = points[j][1];

                int dist = abs(x2 - x1) + abs(y2 - y1);

                edges.push_back({i, j, dist});
            }
        }

        int sum = 0;

        vector<int> parent(n);
        vector<int> rank(n, 0);

        for (int i = 0; i < n; i++) {
            parent[i] = i;
        }

        // Sort edges by weight
        auto lambda = [](const vector<int>& v1, const vector<int>& v2) {
            return v1[2] < v2[2];
        };

        sort(edges.begin(), edges.end(), lambda);

        // Kruskal's Algorithm
        int selectedEdges = 0;
        for (auto& temp : edges) {

            int u = temp[0];
            int v = temp[1];
            int wt = temp[2];

            if (Union(parent, rank, u, v)) {
                sum += wt;
                selectedEdges++;

                if (selectedEdges == n - 1) {
                    break;
                }
            }
        }

        return sum;//TC:O(n^2 log n) || SC:O(n^2)
    }
};
````
