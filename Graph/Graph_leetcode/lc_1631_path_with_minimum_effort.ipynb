# LeetCode 1631 : Path with Minimum Effort

## Problem
- Given a `heights` grid, move between cells in the four directions.
- The effort of a path is the **maximum absolute height difference** between adjacent cells on that path.
- Return the minimum effort needed to travel from the top-left cell to the bottom-right cell.

## Approach: Dijkstra + Min-Heap
- Treat each cell as a graph node, with edges to its valid four-direction neighbors.
- The edge cost is `abs(heights[x][y] - heights[X][Y])`.
- `result[x][y]` stores the smallest effort found so far to reach cell `(x, y)`; initialize all cells to `INT_MAX` except `result[0][0] = 0`.
- The min-priority queue stores `{effort, {row, column}}`, so the cell with the smallest known effort is processed first.
- Unlike ordinary Dijkstra, do not add edge costs. The effort of an extended path is `max(current effort, next edge cost)` because a path is measured by its largest step.

## Steps and Code Logic
1. Set the starting cell's effort to `0` and push `{0, {0, 0}}` into the priority queue.
2. Pop the cell with the smallest effort and read its coordinates.
3. If it is the bottom-right cell, return its effort.
4. Skip the entry if its effort is greater than `result[x][y]`; a better route to that cell was already found.
5. For each valid neighbor, calculate `newDiff = abs(heights[x][y] - heights[X][Y])` and `maxDiff = max(diff, newDiff)`.
6. If `maxDiff` improves `result[X][Y]`, update it and push the neighbor into the priority queue.
7. Continue until the destination is reached.

## Why It Works / Details to Remember
- Every route to a cell has a bottleneck effort: the largest height difference used along that route.
- Extending a route can only keep or increase its effort, so the relaxation is `max(current effort, edge difference)`.
- The min-heap expands the route with the smallest current bottleneck first. Therefore, when the destination is removed from the heap, its effort is the minimum possible.
- The stale-entry check avoids reprocessing a cell using an outdated, larger effort.
- `directions` contains only the four adjacent moves, as required by the problem.

## Complexity
- Let `R` be the number of rows and `C` the number of columns.
- Time: `O(R * C log(R * C))` with a min-priority queue.
- Space: `O(R * C)` for the `result` matrix and priority queue.

## Code

```cpp
class Solution {

public:
    vector<vector<int>> directions = {{1, 0}, {0, 1}, {-1, 0}, {0, -1}};
    typedef pair<int, pair<int, int>> P; //{diff , { i , j}}
    int minimumEffortPath(vector<vector<int>>& heights) {
        int n = heights.size();
        int m = heights[0].size();

        vector<vector<int>> result(n, vector<int>(m, INT_MAX));
        priority_queue<P, vector<P>, greater<P>> pq;

        result[0][0] = 0;
        pq.push({0, {0, 0}});

        while (!pq.empty()) {
            int diff = pq.top().first;
            auto cord = pq.top().second;
            int x = cord.first;
            int y = cord.second;
            pq.pop();

            if (x == n - 1 && y == m - 1)
                return diff;
            if (diff > result[x][y])
                continue;
            for (auto& dir : directions) {
                int X = x + dir[0];
                int Y = y + dir[1];

                if (X >= 0 && Y >= 0 && X < n && Y < m) {
                    int newDiff =
                        abs(heights[x][y] - heights[X][Y]);
                    int maxDiff =
                            max(newDiff, diff);
                    if (maxDiff < result[X][Y]) {
                        result[X][Y] = maxDiff;
                        pq.push({maxDiff, {X, Y}});
                    }
                }
            }
        }

        return 0;
    }
};
```
