# Leetcode 3286 : Find Safe Walk in Grid

## Problem Statement

You are given an `m x n` binary grid and an integer `health`.

- Start at the top-left cell `(0, 0)`.
- Move up, down, left, or right.
- Entering a cell containing `1` reduces health by `1`.
- Reaching the bottom-right cell `(n-1, m-1)` requires at least `1` health remaining.
- Return `true` if a safe path exists; otherwise, return `false`.

## Approach

Use **0-1 BFS** because each move has a cost of `0` or `1`.

- Moving into a cell containing `0` has cost `0`.
- Moving into a cell containing `1` has cost `1`.

Maintain `res[i][j]`, the minimum damage required to reach `(i, j)`.

For a move from `(x, y)` to `(X, Y)`:

```cpp
newDamage = res[x][y] + grid[X][Y]
```

- Insert the neighbour at the front if `grid[X][Y] == 0`.
- Insert the neighbour at the back if `grid[X][Y] == 1`.

Finally, return `true` when:

```cpp
health - res[n - 1][m - 1] >= 1
```

## Algorithm

1. Initialize every entry of `res` with `INT_MAX`.
2. Set `res[0][0] = grid[0][0]` and push `(0, 0)` into the deque.
3. Remove a cell from the front of the deque.
4. Explore all four valid neighbours.
5. Calculate `newDamage = res[x][y] + grid[X][Y]`.
6. If `newDamage` is smaller, update `res[X][Y]`.
7. Add the neighbour to the front for cost `0` or the back for cost `1`.
8. Repeat until the deque is empty.
9. Check whether the destination can be reached with at least `1` health remaining.

## Core Logic

```cpp
for (auto& dir : directions) {
    int X = x + dir[0];
    int Y = y + dir[1];

    if (X >= 0 && Y >= 0 && X < n && Y < m &&
        res[x][y] + grid[X][Y] < res[X][Y]) {

        res[X][Y] = res[x][y] + grid[X][Y];

        if (grid[X][Y] == 0)
            dq.push_front({X, Y});
        else
            dq.push_back({X, Y});
    }
}
```

## Dry Run

Consider:

```text
grid = {
    {0, 1, 0},
    {1, 1, 0},
    {1, 0, 0}
}

health = 4
```

Initial state:

```text
res = {
    {0, INF, INF},
    {INF, INF, INF},
    {INF, INF, INF}
}
```

1. Process `(0, 0)`:
   - `(0, 1)` gets damage `1` and is added at the back.
   - `(1, 0)` gets damage `1` and is added at the back.

2. Process `(0, 1)`:
   - `(0, 2)` gets damage `1` and is added at the front.

3. Process `(0, 2)`:
   - `(1, 2)` gets damage `1` and is added at the front.

4. Process `(1, 2)`:
   - `(2, 2)` gets damage `1` and is added at the front.

Minimum damage to reach the destination is `1`:

```text
remainingHealth = 4 - 1 = 3
```

Since `3 >= 1`, the function returns `true`.

## Key Points

- This is a shortest-path problem with edge weights `0` and `1`.
- `res` stores the minimum damage for every visited cell.
- `grid[X][Y]` determines the deque insertion position.
- Moving into `0` has no cost and uses `push_front`.
- Moving into `1` costs `1` and uses `push_back`.
- The strict condition `health - damage >= 1` ensures at least one unit remains.
- **Time Complexity:** `O(m * n)`
- **Auxiliary Space:** `O(m * n)`

````python
class Solution {
public:
    vector<vector<int>> directions = {{1, 0}, {0, 1}, {-1, 0}, {0, -1}};
    bool findSafeWalk(vector<vector<int>>& grid, int health) {
        int n = grid.size();
        int m = grid[0].size();
        vector<vector<int>> res(n, vector<int>(m, INT_MAX));
        deque<pair<int, int>> dq;
        res[0][0] = grid[0][0];
        dq.push_front({0, 0});
        while (!dq.empty()) {
            auto [x, y] = dq.front();
            dq.pop_front();
            for (auto& dir : directions) {
                int X = x + dir[0];
                int Y = y + dir[1];
                if (X >= 0 && Y >= 0 && X <= n - 1 && Y <= m - 1 &&
                    res[x][y] + grid[X][Y] < res[X][Y]) {

                    res[X][Y] = res[x][y] + grid[X][Y];
                    if (grid[X][Y] == 0)
                        dq.push_front({X, Y});
                    else
                        dq.push_back({X, Y});
                }
            }
        }
        return health - res[n - 1][m - 1] >= 1;//TC:O(m*n) || S:O(m*n)
    }
};
````
