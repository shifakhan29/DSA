# 994. Rotting Oranges

## Approach

Use **Multi-Source BFS** because all initially rotten oranges spread simultaneously. Each BFS level represents one minute.

## Algorithm

1. Count the fresh oranges and push every rotten orange into the queue.
2. If no fresh oranges exist, return `0`.
3. Process the queue one level at a time.
4. For every rotten orange, visit its four adjacent positions.
5. If the adjacent position is fresh:
   - Change it to rotten.
   - Add it to the queue.
   - Decrease `fresh`.
6. Increase `time` after completing each level.
7. If `fresh == 0`, return `time - 1`; otherwise, return `-1`.

## Key Points

- `0` = empty cell
- `1` = fresh orange
- `2` = rotten orange
- A fresh orange is marked as rotten when it enters the queue, so it is not processed again.
- `queue.size()` separates each minute during the BFS traversal.
- Changing a fresh orange to rotten is the same as marking it visited.

## Main Logic

```cpp
for(auto &dir : directions) {
    int X = x + dir[0];
    int Y = y + dir[1];

    if(X >= 0 && Y >= 0 &&
       X < m && Y < n &&
       grid[X][Y] == 1) {
        grid[X][Y] = 2;
        q.push({X, Y});
        fresh--;
    }
}
```

**Minimum time:** each BFS level adds one minute.

- `time - 1` → all fresh oranges became rotten
- `-1` → some fresh oranges remain unreachable

**Time Complexity:** `O(m × n)`  
**Space Complexity:** `O(m × n)`


````python
class Solution {
public:
    typedef pair<int,int>P;
    vector<vector<int>>directions={{1,0} ,{0,1}, {-1,0} ,{0,-1}};
    int orangesRotting(vector<vector<int>>& grid) {
        int m = grid.size();
        int n = grid[0].size();
        queue<P>q;
        int fresh = 0;
        for(int i = 0 ; i< m ; i++){
            for(int j = 0 ; j<n ; j++){
                if(grid[i][j] == 2){
                    q.push({i,j});
                }
                if(grid[i][j]==1){
                    fresh++;
                }
            }
        }
        if(fresh == 0)
        return 0;

        int time = 0;

        while(!q.empty()){
            int s = q.size();
            for(int i = 0 ; i<s ; i++){

            P curr = q.front();
            q.pop();

            int x = curr.first;
            int y = curr.second;
            
            for(auto &dir : directions){
                int X = x + dir[0];
                int Y = y + dir[1];
                if(X>=0 && Y>=0 && X<= m-1 && Y <= n-1 && grid[X][Y] ==1){
                    grid[X][Y] = 2;
                    q.push({X , Y});
                    fresh--;
                }
            }
        }
        time++;
        }
        return (fresh == 0) ?  time-1 : -1;
    }
};
````
