# LeetCode 1319: Number of Operations to Make Network Connected

## Approach: DSU (Union by Rank + Path Compression)

- Initially, every node belongs to a separate component.
- If the parents of two nodes are different, perform **Union**.
- After successful Union, decrease `components` by 1.
- If both nodes have the same parent, the edge is redundant (extra edge).
- To connect all components, we need `components - 1` edges.
- If total edges `< n - 1`, return `-1`.

## Algorithm

1. Check if `connections.size() < n - 1`. If yes, return `-1`.
2. Initialize `parent` and `rank` arrays.
3. Initially, `parent[i] = i`.
4. Traverse all connections:
   - Find parents of both nodes.
   - If parents are different, perform Union and decrement `components`.
   - Otherwise, the edge is redundant.
5. Return `components - 1`.

## Core Logic

- Different parents → Union → `components--`
- Same parent → Redundant edge.
- Required operations = `components - 1`.

## Complexity Analysis

- **TC:** `O(N + E × α(N))`
- **SC:** `O(N)`

Where:
- `N` = Number of nodes.
- `E` = Number of edges.
- `α(N)` = Inverse Ackermann Function (nearly constant).

## Remember

- Successful Union reduces the number of components by 1.
- Same parent means both nodes are already connected.
- Minimum edges required to connect `components` = `components - 1`.
- DSU uses Path Compression + Union by Rank.

## Code

```cpp
class Solution {
public:
    int find(int u, vector<int>& parent) {
        if (parent[u] == u)
            return u;

        return parent[u] = find(parent[u], parent);
    }

    void Union(int u, int v, vector<int>& parent, vector<int>& rank) {
        int U = find(u, parent); // U -> parent of u
        int V = find(v, parent); // V -> parent of v

        if (U == V) { // If both have the same parent, return
            return;
        }

        if (rank[U] > rank[V]) { // Make the node with higher rank the parent
            parent[V] = U;
        }
        else if (rank[U] < rank[V]) {
            parent[U] = V;
        }
        else { // If both ranks are equal, make one the parent and increment its rank
            parent[U] = V;
            rank[V]++;
        }
    }

    int makeConnected(int n, vector<vector<int>>& connections) {

        if (connections.size() < n - 1) // n -> nodes, connections -> edges
            return -1;

        vector<int> parent(n);
        vector<int> rank(n, 0);

        for (int i = 0; i < n; i++) {
            parent[i] = i;
        }

        int components = n;

        for (int i = 0; i < connections.size(); i++) {
            int a = connections[i][0];
            int b = connections[i][1];

            if (find(a, parent) != find(b, parent)) { // If both have different parents
                Union(a, b, parent, rank); // Perform Union
                components--; // Decrease the number of connected components
            }
        }

        return components - 1; // Required edges = components - 1
    }
};

// TC: O(N + E * α(N)) || SC: O(N)
// E -> Edges || N -> Nodes
// α(N) -> Inverse Ackermann Function (nearly constant)
// Due to Path Compression and Union by Rank.
```
