# 1. Shortest Path in an Undirected Graph using Dijkstra's Algorithm

## Approach
- Build an adjacency list where each edge is stored as `{neighbor, weight}`.
- Maintain a `dist[]` array to store the shortest known distance from the source.
- Maintain a `parent[]` array to track the previous node for path reconstruction.
- Use a min-heap (`priority_queue`) so the node with the smallest distance is processed first.
- This is a Greedy + Relaxation approach and works for graphs with non-negative weights.

## Algorithm
1. Initialize all distances with `INT_MAX` and set `dist[src] = 0`.
2. Push `{0, src}` into the min-heap.
3. While the heap is not empty:
   - Pop the node with the minimum distance.
   - If `d > dist[currNode]`, skip it because it is an outdated entry.
   - Traverse all adjacent nodes `(nextNode, wt)`.
   - If `d + wt < dist[nextNode]`, update the distance and parent.
4. After processing all nodes, `dist[]` contains the shortest distances from the source.
5. Reconstruct the path from `dest` using the `parent[]` array.
6. If `dist[dest] == INT_MAX`, then no valid path exists.

## Core Logic

```cpp
if (d + wt < dist[nextNode]) {
    dist[nextNode] = d + wt;
    parent[nextNode] = currNode;
    pq.push({d + wt, nextNode});
}
```

- `dist[]` ? stores the shortest distance from the source.
- `parent[]` ? stores the previous node used to reach the current node.
- `pq` ? min-heap storing `{distance, node}` pairs.
- We only update a node when a shorter path is found.

## Path Reconstruction

```cpp
vector<int> path;
int node = dest;

if (dist[dest] == INT_MAX) return {-1};

while (node != src) {
    path.push_back(node);
    node = parent[node];
}

path.push_back(src);
reverse(path.begin(), path.end());
```

## Complexity Analysis

| Complexity | Value |
|-----------|-------|
| Time Complexity | `O((V + E) log V)` |
| Space Complexity | `O(V + E)` |

- `V` = number of vertices
- `E` = number of edges

## Remember

- Dijkstra works only with **non-negative edge weights**.
- Always process the vertex with the minimum distance first.
- It is optimal for single-source shortest path problems in weighted graphs.
- The parent array helps reconstruct the exact path.

## Code

```cpp
vector<int> shortestPath(vector<vector<pair<int, int>>> &adj, int src, int dest) {
    int n = adj.size();
    vector<int> dist(n, INT_MAX);
    vector<int> parent(n, -1);
    priority_queue<pair<int, int>, vector<pair<int, int>>, greater<pair<int, int>>> pq;

    dist[src] = 0;
    parent[src] = src;
    pq.push({0, src});

    while (!pq.empty()) {
        auto [d, node] = pq.top();
        pq.pop();

        if (d > dist[node]) continue;

        for (auto &[nextNode, wt] : adj[node]) {
            if (d + wt < dist[nextNode]) {
                dist[nextNode] = d + wt;
                parent[nextNode] = node;
                pq.push({d + wt, nextNode});
            }
        }
    }

    if (dist[dest] == INT_MAX) {
        return {-1};
    }

    vector<int> path;
    int node = dest;

    while (node != src) {
        path.push_back(node);
        node = parent[node];
    }

    path.push_back(src);
    reverse(path.begin(), path.end());
    return path;
}
```
