# 1. Dijkstra's Algorithm (Shortest Path using Priority Queue)

## Approach
- Used to find the **shortest distance from a single source node to all other nodes** in a weighted graph.
- Works with **non-negative edge weights**.
- Uses a **Priority Queue (Min-Heap)** to process the node having the minimum distance first.
- Uses **Greedy Approach + Relaxation**.

## Algorithm
1. Create a `dist[]` array and initialize all distances with `INT_MAX`.
2. Set the source node's distance to `0`.
3. Create a min-heap priority queue storing `{distance, node}`.
4. Push `{0, source}` into the priority queue.
5. While the priority queue is not empty:
   - Extract the node with the minimum distance.
   - Traverse all its adjacent nodes.
   - Calculate the new distance: `currentDist + wt`.
   - If `newDist < dist[v]`:
     - Update `dist[v]`.
     - Push `{newDist, v}` into the priority queue.
6. When the priority queue becomes empty, `dist[]` contains the shortest distances from the source.

## Core Logic

**Relaxation:**
```cpp
if (currentDist + wt < dist[v]) {
    dist[v] = currentDist + wt;
    pq.push({dist[v], v});
}
```
- `dist[]` → Stores minimum distance from source.
- `pq` → Min-heap storing `{distance, node}`.
- `wt` → Edge weight.
- Update only when a shorter path is found.

## Complexity Analysis

| Complexity       | Value              |
|------------------|--------------------|
| Time Complexity  | `O((V + E) log V)` |
| Space Complexity | `O(V + E)`         |

- `V` → Number of vertices.
- `E` → Number of edges.

## Remember

- **Dijkstra = Greedy + Min-Heap + Relaxation**
- Always process the node with minimum distance first.
- Works only with non-negative edge weights.
- Can be used for directed and undirected weighted graphs.
- Does not work correctly with negative edge weights.

## Code
```cpp
class Solution {
public:
    vector<int> dijkstra(vector<vector<vector<int>>> &adj, int src) {

        priority_queue<pair<int, int>, vector<pair<int, int>>,
        greater<pair<int, int>>> pq; // Min-heap: {distance, node}

        int n = adj.size();

        vector<int> ans(n, INT_MAX); // Initialize all distances to infinity

        ans[src] = 0; // Distance from source to itself is 0

        pq.push({0, src}); // Push {distance, source node} into min-heap

        while (!pq.empty()) {

            int d = pq.top().first;  // Get minimum distance
            int node = pq.top().second; // Get corresponding node

            pq.pop(); // Remove the minimum-distance pair

            if( d < ans[node]){
                continue;//skip outdated entries
            }

            for (auto &v : adj[node]) { // Traverse all adjacent nodes

                int adjNode = v[0]; // Get adjacent node
                int wt = v[1];      // Get edge weight

                // Relaxation: Check if a shorter path is found
                if (d + wt < ans[adjNode]) {

                    ans[adjNode] = d + wt; // Update shortest distance

                    pq.push({d + wt, adjNode}); // Push updated {distance, node}
                }
            }
        }

        return ans; // Return shortest distances from source to all nodes
    }
};

```

# 2. Dijkstra Algorithm using Set

## Approach
- Initialize the distance of every node to `INT_MAX` (infinity).
- Set the source node's distance to `0`.
- Use an **ordered set** to store `{distance, node}` pairs.
- `set` automatically maintains elements in sorted order.
- Always extract the node with the smallest distance using `*st.begin()`.
- Relax all adjacent edges:
  - If `d + wt < res[adjNode]`, a shorter path is found.
  - Erase the previous `{distance, node}` pair from the set.
  - Update the distance and insert the new pair.

## Algorithm
1. Create a distance array initialized with `INT_MAX`.
2. Set `res[src] = 0`.
3. Insert `{0, src}` into the set.
4. While the set is not empty:
   - Extract the smallest `{distance, node}` pair.
   - Remove it from the set.
   - Traverse all adjacent nodes.
   - If a shorter distance is found:
     - Erase the old pair from the set (if it exists).
     - Update the distance.
     - Insert the updated pair.
5. Return the distance array.

## Remember

- `set` stores unique pairs in sorted order.
- `*st.begin()` → Gives the smallest distance.
- `st.erase({oldDistance, node})` → Removes the previous entry.
- `st.insert({newDistance, node})` → Inserts the updated entry.
- Unlike a priority queue, a set allows **direct deletion of outdated distances**.
- Dijkstra works only with **non-negative edge weights**.

## Core Logic

```cpp
auto it = *st.begin();

int d = it.first;
int node = it.second;

st.erase(st.begin());

for (auto &v : adj[node]) {

    int adjNode = v[0];
    int wt = v[1];

    if (d + wt < res[adjNode]) {

        if (res[adjNode] != INT_MAX)
            st.erase({res[adjNode], adjNode});

        res[adjNode] = d + wt;
        st.insert({d + wt, adjNode});
    }
}
```

## Complexity Analysis

- **Time Complexity:** `O((V + E) log V)` 
  - Set insertion, deletion and minimum-element extraction take `O(log V)`.
  - Each edge is considered during relaxation.

- **Space Complexity:** `O(V + E)`
  - Adjacency list: `O(V + E)`
  - Distance array and set: `O(V)` each.


## Complete Code(GFG)

```cpp
class Solution {
public:
    vector<int> dijkstra(vector<vector<vector<int>>> &adj, int src) {

        set<pair<int, int>> st;

        int n = adj.size();
        vector<int> res(n, INT_MAX);

        res[src] = 0;
        st.insert({0, src});

        while (!st.empty()) {

            auto it = *st.begin();

            int d = it.first;
            int node = it.second;

            st.erase(st.begin());

            for (auto &v : adj[node]) {

                int adjNode = v[0];
                int wt = v[1];

                if (d + wt < res[adjNode]) {

                    if (res[adjNode] != INT_MAX) {
                        st.erase({res[adjNode], adjNode});
                    }

                    res[adjNode] = d + wt;
                    st.insert({d + wt, adjNode});
                }
            }
        }

        return res;
    }
};
```
