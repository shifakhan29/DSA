# Bipartite Graph

## 1. Definition

A **Bipartite Graph** is a graph whose vertices can be divided into two disjoint sets such that no two adjacent vertices belong to the same set.

- Uses **two colors** to color vertices.
- Every edge must connect vertices of different colors.
- No adjacent vertices should have the same color.

## 2. Important Property

A graph is bipartite **if and only if it contains no odd-length cycle**.

| Cycle | Bipartite? |
|---|---|
| 3 vertices (Triangle) | ❌ No |
| 4 vertices | ✅ Yes |
| 5 vertices | ❌ No |
| 6 vertices | ✅ Yes |

**Remember:**
- Odd cycle → Not Bipartite.
- Even cycle → Bipartite.
- A graph without cycles is also bipartite.

## 3. Graph Coloring

Use two colors:

- `0` → First color
- `1` → Second color
- `-1` → Unvisited / Uncolored

Opposite color formula:

```cpp
1 - color[u]
```

## 4. Bipartite Graph Using BFS

### Algorithm

1. Initialize the color array with `-1`.
2. Traverse every vertex to handle disconnected components.
3. If a vertex is uncolored:
   - Assign color `0`.
   - Push it into the queue.
4. While the queue is not empty:
   - Extract the front vertex `u`.
   - Traverse all adjacent vertices `v`.
   - If `v` is uncolored:
     - Assign opposite color: `1 - color[u]`.
     - Push `v` into the queue.
   - Else if `color[v] == color[u]`, return `false`.
5. If all components are processed, return `true`.

### C++ Code

```cpp
// Check Bipartite Graph using BFS
// TC: O(V + E) | SC: O(V)

class Solution {
public:

    bool bfs(vector<vector<int>>& adj, vector<int>& color, int clr, int u) {

        color[u] = clr; // Assign starting color

        queue<int> q;
        q.push(u);

        while (!q.empty()) {

            int u = q.front();
            q.pop();

            for (int &v : adj[u]) {

                // Adjacent vertices cannot have the same color
                if (color[v] == color[u])
                    return false;

                // Visit uncolored neighbour with opposite color
                else if (color[v] == -1) {
                    color[v] = 1 - color[u];
                    q.push(v);
                }
            }
        }

        return true;
    }

    bool isBipartite(vector<vector<int>>& adj) {

        int n = adj.size();
        int clr = 0; // Initial color

        vector<int> color(n, -1); // -1 = uncolored

        // Check all components (handles disconnected graphs)
        for (int i = 0; i < n; i++) {

            if (color[i] == -1) {

                if (!bfs(adj, color, clr, i))
                    return false;
            }
        }

        return true;
    }
};
```
## 5. Bipartite Graph Using DFS

### Algorithm

1. Initialize all vertices with color `-1`.
2. Traverse every vertex to handle disconnected components.
3. If a vertex is uncolored, start DFS with color `0`.
4. Mark the current vertex with the given color.
5. Traverse all adjacent vertices:
   - If uncolored, recursively call DFS with opposite color (`1 - clr`).
   - If already colored with the same color, return `false`.
6. If every component passes, return `true`.

### C++ Code
```cpp
// Check Bipartite Graph using DFS
// TC: O(V + E) | SC: O(V)

class Solution {
public:

    bool dfs(vector<vector<int>>& adj, vector<int>& color, int u, int clr) {

        color[u] = clr; // Assign current color to vertex

        for (int &v : adj[u]) {

            // Adjacent vertices cannot have the same color
            if (color[v] == color[u])
                return false;

            // Visit uncolored neighbour with opposite color
            if (color[v] == -1) {
                if (!dfs(adj, color, v, 1 - clr))
                    return false;
            }
        }

        return true;
    }

    bool isBipartite(vector<vector<int>>& adj) {

        int n = adj.size();
        int clr = 0; // Start coloring with 0

        vector<int> color(n, -1); // -1 = uncolored

        // Check all components (including disconnected graphs)
        for (int i = 0; i < n; i++) {

            if (color[i] == -1 && !dfs(adj, color, i, clr))
                return false;
        }

        return true;
    }
};
```

## 6. Complexity Analysis

| Approach | Time Complexity | Space Complexity |
|---|---|---|
| BFS | `O(V + E)` | `O(V)` |
| DFS | `O(V + E)` | `O(V)` |

Where:
- `V` = Number of vertices
- `E` = Number of edges

## 7. Important Conditions

```cpp
// Assign opposite color
color[v] = 1 - color[u];

// Same adjacent colors → Not Bipartite
if (color[v] == color[u])
    return false;
```

## 8. BFS vs DFS

| BFS | DFS |
|---|---|
| Uses Queue | Uses Recursion / Stack |
| Level-wise traversal | Depth-wise traversal |
| Assigns opposite color | Assigns opposite color |
| Detects coloring conflicts | Detects coloring conflicts |

## Quick Revision

- Bipartite Graph → Two-colorable graph.
- Colors → `0` and `1`.
- Initial color → `-1`.
- Opposite color → `1 - color[u]`.
- Same adjacent colors → Not Bipartite.
- Odd cycle → Not Bipartite.
- Even cycle → Bipartite.
- Disconnected graph → Check every component.

**Key Concept:** Bipartite Graph = Graph Coloring + BFS/DFS.
